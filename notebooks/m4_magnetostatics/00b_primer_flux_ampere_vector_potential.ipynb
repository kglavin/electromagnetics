## 2. The ideas

**Two local laws.** In module 2 the electrostatic field turned out to obey two statements at every point: its divergence is the local charge density over $\epsilon_0$, and its curl is zero. Those two facts, plus boundary conditions, fixed $\mathbf{E}$ completely. The magnetic field of steady currents obeys two statements of the same shape with the roles swapped. Its divergence is zero everywhere, $\nabla\cdot\mathbf{B} = 0$: there is no magnetic charge, nothing for a field line to start on or end on, so every field line closes on itself or runs off to infinity. Its curl is not zero but is set by the current density, $\nabla\times\mathbf{B} = \mu_0\mathbf{J}$: wherever current flows, the magnetic field circulates around it. Primer 00c of module 1 defined divergence as flux out of a tiny box and curl as circulation around a tiny loop. Those definitions translate the two laws directly: the magnetic flux out of any tiny box is zero, and the circulation of $\mathbf{B}$ around any tiny loop is $\mu_0$ times the current threading it.

**The integral forms.** The divergence theorem and Stokes' theorem (primer M1.0d) turn the tiny boxes and loops into finite ones. Integrate $\nabla\cdot\mathbf{B} = 0$ over any volume and the divergence theorem gives $\oint\mathbf{B}\cdot d\mathbf{S} = 0$ over its closed surface: the total magnetic flux through any closed surface is zero, whatever currents are inside or outside. Integrate $\nabla\times\mathbf{B} = \mu_0\mathbf{J}$ over any open surface and Stokes' theorem gives $\oint\mathbf{B}\cdot d\mathbf{l} = \mu_0 I_{\text{enc}}$ around its rim, where $I_{\text{enc}}$ is the current passing through the surface. That is Ampere's law. It is to magnetostatics what Gauss' law is to electrostatics: always true, and a shortcut to the field whenever the geometry has enough symmetry to pull $|\mathbf{B}|$ out of the integral.

**Where the laws come from.** Both can be derived from the Biot-Savart law, which is the experimental input, in the same way Gauss' law was derived from Coulomb's law. The divergence of each Biot-Savart element $d\mathbf{l}\times\hat{\mathbf{R}}/R^2$ vanishes identically, which is a two-line calculation done in section 3.1. The curl is harder, and the book's section 12.4 gives the proof; here the path is the one the course takes in practice. Check Ampere's law for the straight wire, where the circulation of $\mu_0I/2\pi r$ around a circle gives $\mu_0 I$ for every radius, then accept the general statement and verify it numerically on fields assembled from thousands of segments, including loops that the symmetry method could never handle.

**The symmetry method.** Ampere's law gives $\mathbf{B}$ in three lines when a loop can be found on which $\mathbf{B}$ is either parallel to the loop with constant magnitude, or perpendicular to it. Four geometries qualify: anything with the symmetry of an infinite straight line (wires, cables), the infinite solenoid, the toroid, and the infinite flat sheet of current. Each is a steady-current system that looks the same under some translation or rotation, and the argument is always: the symmetry forbids certain components, the loop is chosen along the one that survives, and $|\mathbf{B}|$ times the loop length equals $\mu_0 I_{\text{enc}}$. Section 3.3 works all four. For everything else, a single circular loop above all, Ampere's law still holds but gives no shortcut, and the field must be summed from Biot-Savart, which is what notebook 01 does.

**The vector potential.** Because $\nabla\cdot\mathbf{B} = 0$ everywhere, and because the divergence of a curl is always zero (primer M1.0c, section 3.6), $\mathbf{B}$ can always be written as the curl of some vector field $\mathbf{A}$. This is the magnetic twin of writing $\mathbf{E} = -\nabla V$, which was possible because $\nabla\times\mathbf{E} = 0$. Like $V$, the vector potential is not unique: adding the gradient of any scalar leaves $\mathbf{B}$ unchanged. Like $V$, it satisfies a Poisson equation, $\nabla^2\mathbf{A} = -\mu_0\mathbf{J}$, whose solution looks exactly like the potential of a charge distribution with $\rho/\epsilon_0$ replaced by $\mu_0\mathbf{J}$. And like $V$, it is often easier to compute than the field, because each segment contributes $d\mathbf{l}/R$ instead of a cross product over $R^2$. Section 3.5 computes $\mathbf{A}$ of a loop on a grid, takes its curl, and recovers the Biot-Savart field.

**The parallel.** Section 3.4 lays the two theories side by side. Charge is the source of $\mathbf{E}$ and makes it diverge; current is the source of $\mathbf{B}$ and makes it curl. The static electric field has a scalar potential because it does not curl; the magnetic field has a vector potential because it does not diverge. The symmetry method applies to the same list of shapes. The constants $\epsilon_0$ and $\mu_0$ combine into $1/\sqrt{\epsilon_0\mu_0} = c$, which is the first hint, long before module 6, that the two fields are one subject.

**What the main notebooks do.** Notebook 01 computes $\mathbf{B}$ of loops and a Helmholtz pair by Biot-Savart and checks the on-axis closed form. Notebook 02 builds a solenoid, compares its interior field with $\mu_0 nI$, and tests Ampere's law and $\nabla\cdot\mathbf{B} = 0$ on a grid using the line-integral and box-flux routines of module 2. Notebook 04 uses $\nabla\times\mathbf{B} = \mu_0\mathbf{J}$ in reverse: given a magnetization, it finds the bound currents that would produce the same field, which needs primer 00c. Everything those notebooks rely on is derived below.

### 3.1 Magnetic flux and $\nabla\cdot\mathbf{B} = 0$

The flux of $\mathbf{B}$ through a surface $S$ is defined exactly as electric flux was in primer M1.0b,
$$\Phi_B = \int_S\mathbf{B}\cdot d\mathbf{S},$$
the sum over small patches of the normal component of $\mathbf{B}$ times the patch area. Its unit is the weber, $1\,\mathrm{Wb} = 1\,\mathrm{T\,m^2}$, and the tesla is sometimes read backwards as one weber per square metre. The flux through a loop of wire is the quantity that matters in module 6: a change in it drives a current (Faraday's law, chapter 14). For now flux is used for one statement: through any *closed* surface it is zero,
$$\oint_S\mathbf{B}\cdot d\mathbf{S} = 0.$$

The reason is the picture of field lines. Electric field lines start on positive charge and end on negative charge, and Gauss' law counts how many start inside a closed surface. Magnetic field lines have nowhere to start: there is no magnetic charge. A field line of a current loop threads the loop, comes around the outside and threads it again, closing on itself. Every line that enters a closed surface must leave it, so the net flux is zero. By the divergence theorem that is the same as $\nabla\cdot\mathbf{B} = 0$ at every point.

This is not an extra assumption. It follows from the Biot-Savart law.

**Worked example 1 (the divergence of a Biot-Savart element).** One segment of wire contributes $d\mathbf{B} \propto d\mathbf{l}\times\mathbf{R}/R^3$ with $\mathbf{R} = \mathbf{r} - \mathbf{r}'$ the vector from the segment to the field point, and $d\mathbf{l}$ a constant vector. Use the product rule $\nabla\cdot(\mathbf{a}\times\mathbf{b}) = \mathbf{b}\cdot(\nabla\times\mathbf{a}) - \mathbf{a}\cdot(\nabla\times\mathbf{b})$ with $\mathbf{a} = d\mathbf{l}$ and $\mathbf{b} = \mathbf{R}/R^3$. The first term is zero because $d\mathbf{l}$ is constant. The second is zero because $\mathbf{R}/R^3 = -\nabla(1/R)$ is a gradient, and the curl of a gradient vanishes (primer M1.0c, section 3.6). So $\nabla\cdot d\mathbf{B} = 0$ for every segment, and since the divergence of a sum is the sum of the divergences, $\nabla\cdot\mathbf{B} = 0$ for any current at all. Two lines, no symmetry assumed.

**Worked example 2 (flux of a loop's field through a distant disk).** A loop of radius $R = 0.1\,$m carries $I = 1\,$A, so its dipole moment is $m = I\pi R^2 = 0.0314\,\mathrm{A\,m^2}$ (primer 00a). Far away its field is the dipole field, and on the loop's axis at height $h$ the $z$ component at distance $s$ from the axis is $B_z = \dfrac{\mu_0 m}{4\pi}\dfrac{2h^2 - s^2}{(s^2 + h^2)^{5/2}}$. The flux through a coaxial disk of radius $a$ at height $h$ is $\int_0^a B_z\,2\pi s\,ds$. With $u = s^2 + h^2$ the integral is elementary and gives
$$\Phi_B = \frac{\mu_0 m}{2}\frac{a^2}{(a^2 + h^2)^{3/2}}.$$
For $a = 0.5\,$m and $h = 2\,$m: $\Phi_B = \dfrac{4\pi\times10^{-7}\cdot0.0314}{2}\cdot\dfrac{0.25}{(4.25)^{3/2}} = 1.974\times10^{-8}\cdot\dfrac{0.25}{8.762} = 5.63\times10^{-10}\,$Wb. Now let the disk grow. At $a = 2\,$m the flux is $3.49\times10^{-9}\,$Wb, larger; at $a = 10\,$m it is $1.86\times10^{-9}\,$Wb, smaller again; and as $a\to\infty$ the formula goes to zero like $1/a$. The field lines that go up through the middle of the disk come back down through its outer part, and a large enough disk catches all of the return flux. That is $\nabla\cdot\mathbf{B} = 0$ in one picture.

The code checks the disk flux by sampling the Biot-Savart field on the disk (the dipole formula is an approximation whose error is of order $(R/h)^2$, a few tenths of a percent at $h = 2$), then computes the flux of the same field out of a box using `fields.box_flux` and the divergence on the grid with `fields.divergence`. The box is placed off centre so that the zero cannot come from symmetry alone. The divergence is compared with the size of the field's derivatives, since a divergence is a sum of derivatives and "zero" means small compared with the terms being added; the central differences on this grid cancel to under a percent at typical points, and least well nearest the wire where the field changes fastest.

```python
I, R = 1.0, 0.1
mid, dl = bs.segments(bs.loop(R, 400))
m = I * np.pi * R**2
print(f"dipole moment m = {m:.4f} A m^2")
for a, h in ((0.5, 2.0), (2.0, 2.0), (10.0, 2.0)):
    flux_hand = MU0 * m * a**2 / (2 * (a**2 + h**2) ** 1.5)
    s = (np.arange(400) + 0.5) * a / 400
    ph = np.linspace(0, 2 * np.pi, 32, endpoint=False)
    S, PH = np.meshgrid(s, ph, indexing="ij")
    disk = np.stack([S * np.cos(PH), S * np.sin(PH), np.full_like(S, h)], axis=-1)
    Bz = bs.wire_B(mid, dl, I, disk)[..., 2]
    flux_num = np.sum(Bz * S) * (a / 400) * (2 * np.pi / 32)
    print(f"disk a = {a:4.1f} m at h = {h} m:  dipole formula {flux_hand:.3e} Wb   Biot-Savart {flux_num:.3e} Wb   ratio {flux_num / flux_hand:.4f}")

x, y, z, P = grids.grid3d((-0.4, 0.4), (-0.4, 0.4), (-0.4, 0.4), 33)
h3 = grids.spacing(x, y, z)
B = bs.wire_B(mid, dl, I, P, chunk=128)
lo, hi = (6, 9, 12), (26, 28, 30)                       # an off-centre box of grid indices
flux_box = fields.box_flux(B, h3, lo, hi)
sl = tuple(slice(a, b + 1) for a, b in zip(lo, hi))
unsigned = np.abs(B[sl]).mean() * sum(2 * np.prod([(hi[k] - lo[k]) * h3[k] for k in range(3) if k != ax]) for ax in range(3))
print(f"\nflux of B out of the box: {flux_box:+.2e} Wb   compared with the unsigned flux |B| x area: {unsigned:.2e} Wb   ratio {abs(flux_box) / unsigned:.1e}")
div = fields.divergence(B, *h3)
deriv_scale = np.sqrt(sum(np.gradient(B[..., k], h3[a], axis=a) ** 2 for k in range(3) for a in range(3)))   # size of all nine derivatives
away = np.sqrt((np.hypot(P[..., 0], P[..., 1]) - R) ** 2 + P[..., 2] ** 2) > 0.15
away[:2] = away[-2:] = False; away[:, :2] = away[:, -2:] = False; away[:, :, :2] = away[:, :, -2:] = False
ratio = np.abs(div[away]) / deriv_scale[away]
print(f"|div B| compared with the size of B's derivatives, at points more than 0.15 m from the wire: median {np.median(ratio):.1e}, max {ratio.max():.1e}")
```

### 3.2 Ampere's law

Take the long straight wire of primer 00a, whose field is $B_\phi = \mu_0 I/2\pi r$, circling the wire. Walk once around it on a circle of radius $r$ in the direction of the field. The circulation is $\oint\mathbf{B}\cdot d\mathbf{l} = B_\phi\cdot2\pi r = \mu_0 I$. The radius dropped out: a circle of any size gives the same number, and the number is $\mu_0$ times the current inside. The analogy with the flux of a point charge through any sphere is exact, and so is what comes next: for any closed curve $C$ and any surface $S$ whose rim is $C$,

$$\oint_C\mathbf{B}\cdot d\mathbf{l} = \mu_0 I_{\text{enc}}, \qquad I_{\text{enc}} = \int_S\mathbf{J}\cdot d\mathbf{S}.$$

This is Ampere's law. The sign is fixed by the right-hand rule: curl the fingers of the right hand in the direction of travel around $C$ and the thumb gives the direction in which current counts as positive. Currents that do not pass through $S$ contribute nothing, however close they are, just as charges outside a Gaussian surface contributed nothing to the flux. Shrink the loop to a tiny square and divide by its area, and the definition of curl (primer M1.0c, section 3.4) turns the integral form into the differential one, $\nabla\times\mathbf{B} = \mu_0\mathbf{J}$.

**Worked example 3.** $I = 2\,$A in a long straight wire. The circulation around any loop that encloses it once, in the right-hand sense, is $\mu_0 I = 4\pi\times10^{-7}\cdot2 = 2.513\times10^{-6}\,\mathrm{T\,m}$: a circle of radius $0.1\,$m, a circle of radius $3\,$m, a square, a tilted wobbly loop. A loop that does not enclose the wire gives zero. A loop traversed the other way gives $-\mu_0 I$.

**Worked example 4 (several wires).** Three long wires parallel to $z$ carry $+3\,$A at $(0.3, 0)$, $-1\,$A at $(-0.3, 0.2)$ and $+2\,$A at $(1.5, 0)$, the sign giving the direction along $+z$. A circle of radius 1 around the origin, counter-clockwise seen from $+z$, encloses the first two: $I_{\text{enc}} = 3 - 1 = 2\,$A, circulation $2\mu_0$. The third wire is outside and contributes nothing, although its field is certainly present on the loop.

Two points about the right-hand side.

*Which surface?* The rim $C$ bounds infinitely many surfaces, a flat disk and a bulging dome among them, and the law must give the same $I_{\text{enc}}$ for all of them. It does when the current is steady. Steady currents satisfy $\nabla\cdot\mathbf{J} = 0$ (charge does not pile up anywhere), so the current into the volume between any two such surfaces equals the current out, and the same current crosses both. For time-varying currents this fails and Ampere's law needs an extra term; that is Maxwell's correction in module 6. Everything in module 4 is steady.

*Why not use it to find any field?* The law holds for every loop, but it gives one number per loop, a single integral of $\mathbf{B}$, not the field itself. Only when symmetry tells you that $\mathbf{B}$ is constant along the loop and parallel to it can you divide by the loop length. That is the subject of the next section.

The code computes the circulation of a Biot-Savart wire field with `fields.line_integral` around several loops.

```python
I, L = 2.0, 200.0
mid_w, dl_w = bs.segments(bs.straight_wire([0, 0, -L], [0, 0, L], 20000), closed=False)
B_wire = lambda p: bs.wire_B(mid_w, dl_w, I, p)
t = np.linspace(0, 2 * np.pi, 601)
print(f"mu0 I = {MU0 * I:.4e} T m\n")
for r in (0.1, 0.3, 1.0, 3.0):
    circle = np.stack([r * np.cos(t), r * np.sin(t), 0 * t], axis=-1)
    print(f"circle of radius {r}:            circulation / (mu0 I) = {fields.line_integral(B_wire, circle) / (MU0 * I):+.5f}")
corners = np.array([[1, -1, 0], [1, 1, 0], [-1, 1, 0], [-1, -1, 0.0]])
square = np.vstack([np.linspace(corners[k], corners[(k + 1) % 4], 500, endpoint=False) for k in range(4)] + [corners[:1]])
print(f"square of side 2:                circulation / (mu0 I) = {fields.line_integral(B_wire, square) / (MU0 * I):+.5f}")
print(f"same square, traversed backwards: circulation / (mu0 I) = {fields.line_integral(B_wire, square[::-1]) / (MU0 * I):+.5f}")
wobbly = np.stack([np.cos(t), np.sin(t), 0.5 * np.sin(2 * t)], axis=-1)
print(f"tilted wobbly loop:              circulation / (mu0 I) = {fields.line_integral(B_wire, wobbly) / (MU0 * I):+.5f}")
outside = np.stack([3 + 0.8 * np.cos(t), 0.8 * np.sin(t), 0 * t], axis=-1)
print(f"circle not enclosing the wire:   circulation / (mu0 I) = {fields.line_integral(B_wire, outside) / (MU0 * I):+.2e}")

wires = [((0.3, 0.0), 3.0), ((-0.3, 0.2), -1.0), ((1.5, 0.0), 2.0)]       # (x, y) position and current along +z
def B_three(p):
    return sum(bs.wire_B(mid_w + [wx, wy, 0], dl_w, cur, p) for (wx, wy), cur in wires)
unit_circle = np.stack([np.cos(t), np.sin(t), 0 * t], axis=-1)
print(f"\nthree wires, unit circle:        circulation / mu0 = {fields.line_integral(B_three, unit_circle) / MU0:.4f} A   (enclosed current 3 - 1 = 2 A)")
```

### 3.3 Ampere's law by symmetry

The method is the one from primer M2.0a with the surface replaced by a loop. Each part of the Amperian loop must satisfy one of two conditions:

1. $\mathbf{B}$ is parallel to the loop and $|\mathbf{B}|$ is the same all along that part. Then its contribution is $|\mathbf{B}|\times$ length.
2. $\mathbf{B}$ is perpendicular to the loop. Then its contribution is zero.

Both come from symmetry arguments about the current distribution, never from calculation. The argument has the form used for the line charge in primer M2.0a: list the transformations that leave the currents unchanged, and conclude that the field must be unchanged too, which rules out components and dependences. Four distributions have enough symmetry.

**Worked example 5 (infinite straight wire).** Current $I$ along $z$. Translating along $z$ and rotating about $z$ leave the wire unchanged, so $\mathbf{B}$ depends on $r$ only. Reflecting in a plane containing the wire ($\phi\to-\phi$) reverses the direction of a vector's $\phi$ component but leaves a current along $z$ unchanged; the magnetic field is an axial vector and behaves oppositely to $\mathbf{E}$ under reflection, with the result that $B_r$ and $B_z$ are forbidden and only $B_\phi$ survives. (The Biot-Savart law shows the same thing directly: $d\mathbf{l}\times\mathbf{R}$ with $d\mathbf{l}$ along $z$ is always along $\hat{\boldsymbol\phi}$.) Loop: a circle of radius $r$, condition 1 everywhere. $B_\phi\cdot2\pi r = \mu_0 I$, so $B_\phi = \mu_0 I/2\pi r$. Primer 00a got this from an integral; here it is one line.

**Worked example 6 (thick wire).** A wire of radius $R = 1\,$cm carries $I = 100\,$A spread uniformly, $J = I/\pi R^2 = 3.18\times10^5\,\mathrm{A/m^2}$. Same symmetry, same circular loop. Outside, $I_{\text{enc}} = I$ and $B = \mu_0 I/2\pi r$ as before. Inside, only the current within radius $r$ counts, $I_{\text{enc}} = J\pi r^2 = Ir^2/R^2$, so $B\cdot2\pi r = \mu_0 Ir^2/R^2$ and
$$B = \frac{\mu_0 I\,r}{2\pi R^2}, \qquad r \le R.$$
Linear inside, $1/r$ outside, continuous at the surface where it peaks at $\mu_0 I/2\pi R = 2.0\times10^{-3}\,$T. At $r = 0.5\,$cm and at $r = 2\,$cm the field is the same, $1.0\times10^{-3}\,$T. Compare the uniformly charged ball of primer M2.0a, worked example 4: linear inside, inverse law outside.

**Worked example 7 (coaxial cable).** An inner conductor of radius $a$ carries $I$ along $+z$; an outer shell with inner radius $b$ and outer radius $c$ carries $I$ back along $-z$, uniformly over its cross-section. Four regions. $r < a$: $B = \mu_0 Ir/2\pi a^2$ as in the thick wire. $a < r < b$: $B = \mu_0 I/2\pi r$. $b < r < c$: the enclosed current is $I$ minus the part of the return current inside $r$, $I_{\text{enc}} = I\left(1 - \dfrac{r^2 - b^2}{c^2 - b^2}\right) = I\dfrac{c^2 - r^2}{c^2 - b^2}$, so $B = \dfrac{\mu_0 I}{2\pi r}\dfrac{c^2 - r^2}{c^2 - b^2}$, falling to zero at $r = c$. $r > c$: $I_{\text{enc}} = I - I = 0$ and $B = 0$. The cable's field lives entirely in the gap and the shell, which is why coaxial cables neither radiate nor pick up interference at low frequency. For $I = 5\,$A and $a, b, c = 1, 3, 3.5\,$mm: at $r = 2\,$mm, $B = 2\times10^{-7}\cdot5/0.002 = 5.0\times10^{-4}\,$T.

**Worked example 8 (ideal solenoid).** A long cylinder wound with $n$ turns per unit length carrying $I$. Treat the winding as a cylindrical sheet of azimuthal current $nI$ per unit length. Symmetry: translation along the axis and rotation about it leave it unchanged, so $\mathbf{B}$ depends only on $r$; the current is purely azimuthal, and the Biot-Savart elements $d\mathbf{l}\times\mathbf{R}$ with $d\mathbf{l}$ along $\hat{\boldsymbol\phi}$ never produce a $\phi$ component when summed around the circle, so $\mathbf{B}$ lies in the $r$-$z$ plane. The flux argument of section 3.1 applied to a closed cylinder coaxial with the solenoid shows $B_r = 0$ (the curved surface would otherwise carry net flux). So $\mathbf{B} = B_z(r)\hat{\mathbf{z}}$.

Now two rectangular loops in a plane through the axis, each with two sides parallel to the axis of length $\ell$ and two radial sides (condition 2, zero contribution). First loop entirely outside the winding: it encloses no current, so $B_z(r_1)\ell - B_z(r_2)\ell = 0$ and $B_z$ is the same at every outside radius. Far away the field must vanish, so $B_z = 0$ everywhere outside. Second loop with one axial side inside and one outside: it encloses $n\ell$ turns, current $n\ell I$, and the outside side contributes nothing, so $B_z\ell = \mu_0 n\ell I$:
$$B = \mu_0 nI \quad\text{inside, uniform}, \qquad B = 0 \quad\text{outside}.$$
For $n = 500\,\mathrm{m^{-1}}$ and $I = 2\,$A: $B = 4\pi\times10^{-7}\cdot1000 = 1.257\times10^{-3}\,$T, independent of the solenoid's radius and of where inside you measure. A real solenoid of finite length comes close: on the axis of one with length $L$ and radius $R$ the field is $\dfrac{\mu_0 nI}{2}\left(\dfrac{L/2 + z}{\sqrt{R^2 + (L/2 + z)^2}} + \dfrac{L/2 - z}{\sqrt{R^2 + (L/2 - z)^2}}\right)$ (primer 00a), which is $\mu_0 nI\cdot L/\sqrt{L^2 + 4R^2}$ at the centre and exactly half of $\mu_0 nI$ at each end for a long solenoid.

**Worked example 9 (toroid).** Bend the solenoid into a ring: $N$ turns wound on a doughnut, current $I$. Rotation about the doughnut's axis is the symmetry, so $\mathbf{B}$ depends on $r$ (distance from that axis) and $z$ only, and the same arguments as before leave only $B_\phi$, running around the ring inside the winding. Loop: a circle of radius $r$ about the axis. Inside the winding it is pierced by all $N$ turns: $B_\phi\cdot2\pi r = \mu_0 NI$, so $B_\phi = \mu_0 NI/2\pi r$. It is not uniform; it is stronger at the inner edge. A circle inside the hole or outside the doughnut encloses no net current ($N$ turns go through one way and $N$ come back), so $B = 0$ there. For $N = 500$, $I = 2\,$A: at $r = 0.10\,$m, $B = 2\times10^{-7}\cdot1000/0.1 = 2.0\times10^{-3}\,$T; at $r = 0.12\,$m, $1.67\times10^{-3}\,$T.

**Worked example 10 (infinite sheet of current).** A flat sheet in the $xy$ plane carries a surface current $K$ amperes per metre of width, flowing along $+x$. Translations in $x$ and $y$ leave it unchanged, so $\mathbf{B}$ depends on $z$ only. A Biot-Savart element $d\mathbf{l}\times\mathbf{R}$ with $d\mathbf{l}$ along $x$ has no $x$ component, and summing over the sheet cancels the $z$ components by symmetry (the element at $+y$ cancels the one at $-y$), leaving $B_y$ only. Reflection in the sheet reverses it: $B_y(-z) = -B_y(z)$. Loop: a rectangle in the $yz$ plane straddling the sheet, with sides of length $\ell$ parallel to $y$ at heights $\pm z$ (condition 1) and short sides along $z$ (condition 2). Enclosed current $K\ell$: $2B\ell = \mu_0 K\ell$,
$$B = \frac{\mu_0 K}{2} \quad\text{on each side, parallel to the sheet, perpendicular to the current, independent of distance.}$$
For $K = 1000\,$A/m, $B = 6.28\times10^{-4}\,$T. Compare the charged sheet, $\sigma/2\epsilon_0$. Two parallel sheets with opposite currents give $\mu_0 K$ between them and zero outside, which is the solenoid of worked example 8 seen up close: its wall is a sheet with $K = nI$.

The first code cell below builds a solenoid of 500 turns from a helix and compares its axis field with $\mu_0 nI$ and with the finite-length closed form, then looks off axis: inside the field is uniform to a fraction of a percent, outside it is a few tenths of a percent of the interior value. The small azimuthal field outside is real: a helix carries a net current $I$ along its axis, one wire's worth, and that component matches a straight wire of current $I$. The second cell builds a thick wire from several hundred thin parallel wires and plots its field profile against worked example 6. The bundle's cross-section is a staircase approximation to the circle, so its area is slightly larger than $\pi R^2$; the text under the plot quantifies the effect.

```python
R_s, L_s, turns, I_s = 0.05, 1.0, 500, 2.0
n_density = turns / L_s
helix_pts = bs.helix(R_s, L_s, turns, n_per_turn=60)
mid_s, dl_s = bs.segments(helix_pts, closed=False)
B_ideal = MU0 * n_density * I_s
print(f"solenoid: {turns} turns on {L_s} m, radius {R_s} m, I = {I_s} A, {len(mid_s)} segments.   mu0 n I = {B_ideal:.4e} T")

zz = np.linspace(-0.7, 0.7, 57)
B_axis = bs.wire_B(mid_s, dl_s, I_s, np.stack([0 * zz, 0 * zz, zz], axis=-1))[:, 2]
B_closed = bs.solenoid_axis_B(R_s, L_s, n_density, I_s, zz)
print(f"axis field at the centre / mu0 n I: Biot-Savart {B_axis[28] / B_ideal:.4f},  closed form {B_closed[28] / B_ideal:.4f},  L/sqrt(L^2 + 4R^2) = {L_s / np.sqrt(L_s**2 + 4 * R_s**2):.4f}")
print(f"axis field at the end (z = L/2) / mu0 n I: {B_axis[np.argmin(np.abs(zz - L_s / 2))] / B_ideal:.4f}   (half)")
print(f"max |Biot-Savart - closed form| along the axis / mu0 n I: {np.abs(B_axis - B_closed).max() / B_ideal:.1e}")

rho = np.array([0.0, 0.02, 0.04, 0.045, 0.06, 0.1, 0.2])
B_off = bs.wire_B(mid_s, dl_s, I_s, np.stack([rho, 0 * rho, 0 * rho], axis=-1))
print("\nat z = 0, distance from the axis:", rho, "  (winding at 0.05)")
print("B_z / mu0 n I:                   ", B_off[:, 2] / B_ideal)
print("B_phi / mu0 n I:                 ", B_off[:, 1] / B_ideal)
print("straight wire with current I:    ", np.r_[np.zeros(4), MU0 * I_s / (2 * np.pi * rho[4:]) / B_ideal])

plt.plot(zz, B_axis / B_ideal, "C0.", label="Biot-Savart, 500-turn helix")
plt.plot(zz, B_closed / B_ideal, "k-", lw=1, label="finite-solenoid closed form")
plt.axhline(1, color="grey", ls=":", label="ideal: mu0 n I"); plt.axvline(-L_s / 2, color="grey", ls="--"); plt.axvline(L_s / 2, color="grey", ls="--")
plt.xlabel("z along the axis [m]"); plt.ylabel("B_z / (mu0 n I)"); plt.legend(); plt.title("the ideal solenoid is the long-solenoid limit"); plt.show()
```

```python
R_t, I_t, L_t = 0.01, 100.0, 50.0
n_side = 20
s = -R_t + (np.arange(n_side) + 0.5) * (2 * R_t / n_side)
U, V = np.meshgrid(s, s, indexing="ij")
in_wire = U**2 + V**2 <= R_t**2
cx, cy = U[in_wire], V[in_wire]
# Each thin wire is segmented finely near z = 0, where the field is sampled, and coarsely far away.
u = np.linspace(-1, 1, 1001)
z_graded = L_t * np.sign(u) * np.abs(u) ** 3
mid_1, dl_1 = bs.segments(np.stack([0 * z_graded, 0 * z_graded, z_graded], axis=-1), closed=False)
mid_bundle = np.concatenate([mid_1 + [a, b, 0] for a, b in zip(cx, cy)])
dl_bundle = np.tile(dl_1, (len(cx), 1))
I_thin = I_t / len(cx)
R_eff = np.sqrt(len(cx) / np.pi) * (2 * R_t / n_side)
print(f"thick wire from {len(cx)} thin wires, {len(mid_bundle)} segments; staircase cross-section has effective radius {R_eff / R_t:.4f} R")

def thick_wire_closed_form(r, R, I):
    return np.where(r < R, MU0 * I * r / (2 * np.pi * R**2), MU0 * I / (2 * np.pi * r))

r_prof = np.linspace(0.0005, 0.03, 60)                                   # along y = 0, midway between two rows of thin wires
B_bundle = bs.wire_B(mid_bundle, dl_bundle, I_thin, np.stack([r_prof, 0 * r_prof, 0 * r_prof], axis=-1))[:, 1]
plt.plot(r_prof * 100, B_bundle * 1e3, "C1.", label="bundle of thin wires")
plt.plot(r_prof * 100, thick_wire_closed_form(r_prof, R_t, I_t) * 1e3, "k-", lw=1, label="Ampere: linear inside, 1/r outside")
plt.axvline(R_t * 100, color="grey", ls=":"); plt.xlabel("r [cm]"); plt.ylabel("B [mT]"); plt.legend(); plt.title("thick wire, R = 1 cm, I = 100 A"); plt.show()
for r in (0.005, 0.01, 0.02):
    k = np.argmin(np.abs(r_prof - r))
    print(f"r = {r_prof[k] * 100:.2f} cm: bundle {B_bundle[k] * 1e3:.4f} mT   closed form {thick_wire_closed_form(r_prof[k], R_t, I_t) * 1e3:.4f} mT")
inside = r_prof < 0.8 * R_t
print(f"inside, bundle / closed form with R: {np.mean(B_bundle[inside] / thick_wire_closed_form(r_prof[inside], R_t, I_t)):.4f};  with R_eff: {np.mean(B_bundle[inside] / thick_wire_closed_form(r_prof[inside], R_eff, I_t)):.4f}")
```

The bundle's inside field sits below the closed form with $R$ by about half a percent and within a few tenths of a percent of the closed form with $R_{\text{eff}}$: the current is spread over a slightly larger staircased disk, and the field at a given $r$ inside is $\mu_0 I r/2\pi R_{\text{eff}}^2$. ### 3.4 Electrostatics and magnetostatics side by side

| | electrostatics (modules 1 to 3) | magnetostatics (this module) |
|---|---|---|
| source | charge $q$, density $\rho$ | current $I$, density $\mathbf{J}$ (steady: $\nabla\cdot\mathbf{J} = 0$) |
| elementary law | Coulomb: $d\mathbf{E} = \dfrac{1}{4\pi\epsilon_0}\dfrac{dq\,\hat{\mathbf{R}}}{R^2}$ | Biot-Savart: $d\mathbf{B} = \dfrac{\mu_0}{4\pi}\dfrac{I\,d\mathbf{l}\times\hat{\mathbf{R}}}{R^2}$ |
| force on a test particle | $q\mathbf{E}$, along the field | $q\mathbf{v}\times\mathbf{B}$, perpendicular to field and velocity |
| divergence | $\nabla\cdot\mathbf{E} = \rho/\epsilon_0$ | $\nabla\cdot\mathbf{B} = 0$ |
| curl | $\nabla\times\mathbf{E} = 0$ | $\nabla\times\mathbf{B} = \mu_0\mathbf{J}$ |
| integral law | Gauss: $\oint\mathbf{E}\cdot d\mathbf{S} = Q_{\text{enc}}/\epsilon_0$ | Ampere: $\oint\mathbf{B}\cdot d\mathbf{l} = \mu_0 I_{\text{enc}}$ |
| the other integral | $\oint\mathbf{E}\cdot d\mathbf{l} = 0$ | $\oint\mathbf{B}\cdot d\mathbf{S} = 0$ |
| field lines | start on $+$, end on $-$ | closed, or to infinity |
| potential | scalar $V$, $\mathbf{E} = -\nabla V$ | vector $\mathbf{A}$, $\mathbf{B} = \nabla\times\mathbf{A}$ |
| freedom in the potential | add a constant | add any gradient $\nabla\chi$ |
| potential's equation | $\nabla^2V = -\rho/\epsilon_0$ | $\nabla^2\mathbf{A} = -\mu_0\mathbf{J}$ (Coulomb gauge) |
| potential of a source | $V = \dfrac{1}{4\pi\epsilon_0}\displaystyle\int\dfrac{\rho\,dv}{R}$ | $\mathbf{A} = \dfrac{\mu_0}{4\pi}\displaystyle\int\dfrac{\mathbf{J}\,dv}{R}$ |
| symmetry shortcut | sphere, line/cylinder, plane | line/cylinder, solenoid, toroid, sheet |
| far field of a neutral source | electric dipole $p$: $\propto p/r^3$ | magnetic dipole $m = IA$: $\propto\mu_0 m/r^3$ |
| constant | $\epsilon_0 = 8.854\times10^{-12}\,$F/m | $\mu_0 = 4\pi\times10^{-7}\,$T m/A |

### 3.5 The vector potential

Since $\nabla\cdot\mathbf{B} = 0$ everywhere and $\nabla\cdot(\nabla\times\mathbf{A}) = 0$ for any $\mathbf{A}$, it is natural to ask whether every $\mathbf{B}$ is a curl. It is: a divergence-free field can always be written as
$$\mathbf{B} = \nabla\times\mathbf{A}.$$
The vector potential $\mathbf{A}$ is the magnetic counterpart of $V$, and its existence rests on $\nabla\cdot\mathbf{B} = 0$ the way the existence of $V$ rested on $\nabla\times\mathbf{E} = 0$.

*Gauge freedom.* $\mathbf{A}$ is not unique. For any scalar field $\chi$, $\mathbf{A} + \nabla\chi$ has the same curl, because the curl of a gradient is zero. Just as $V$ was pinned down by choosing $V = 0$ at infinity, $\mathbf{A}$ is pinned down by an extra condition; the usual choice in magnetostatics is $\nabla\cdot\mathbf{A} = 0$, the Coulomb gauge. With it, Ampere's law becomes a Poisson equation. Use the identity $\nabla\times(\nabla\times\mathbf{A}) = \nabla(\nabla\cdot\mathbf{A}) - \nabla^2\mathbf{A}$ (listed in `study/vector_calculus_ref.md`): the first term is zero in the Coulomb gauge, so
$$\nabla^2\mathbf{A} = -\mu_0\mathbf{J},$$
three Poisson equations, one per component, each identical in form to $\nabla^2V = -\rho/\epsilon_0$. Their solution is therefore identical in form to the potential of a charge distribution (primer M1.0e):
$$\mathbf{A}(\mathbf{r}) = \frac{\mu_0}{4\pi}\int\frac{\mathbf{J}(\mathbf{r}')\,dv'}{R}, \qquad\text{for a wire}\quad \mathbf{A} = \frac{\mu_0 I}{4\pi}\int\frac{d\mathbf{l}}{R}.$$
Compare the Biot-Savart law: $\mathbf{B}$ needs $d\mathbf{l}\times\hat{\mathbf{R}}/R^2$ per segment, $\mathbf{A}$ needs only $d\mathbf{l}/R$. No cross product, one power of $R$ less, and the direction of $\mathbf{A}$ simply follows the current. Taking the curl of this integral reproduces Biot-Savart, which is one way to see that the two descriptions agree.

Two uses of $\mathbf{A}$ appear in this course. Flux: by Stokes' theorem, $\Phi_B = \int_S\nabla\times\mathbf{A}\cdot d\mathbf{S} = \oint_C\mathbf{A}\cdot d\mathbf{l}$, so the flux through a loop is the line integral of $\mathbf{A}$ around it, which is how mutual inductance is computed in module 6. And computation: solve three Poisson equations for $\mathbf{A}$ with the sparse solver of module 2, take a curl, and you have $\mathbf{B}$ for a current distribution of any shape.

**Worked example 11 (straight wire).** The current along $z$ makes $\mathbf{A}$ point along $z$, and the integral $\int dz'/R$ over an infinite wire diverges, exactly as the potential of an infinite line charge did (primer M2.0c, worked example 5). Handle it the same way: the answer up to a constant is
$$\mathbf{A} = -\frac{\mu_0 I}{2\pi}\ln\frac{r}{r_0}\,\hat{\mathbf{z}}.$$
Check by taking the curl in cylindrical coordinates (primer M1.0d, section 3.3): for $\mathbf{A} = A_z(r)\hat{\mathbf{z}}$ the only non-zero component is $B_\phi = -\partial A_z/\partial r = \dfrac{\mu_0 I}{2\pi r}$. The line-charge potential and the straight-wire vector potential are the same function because they solve the same Poisson equation with a line source.

**Worked example 12 (uniform field).** A uniform $\mathbf{B}$ has the vector potential $\mathbf{A} = \tfrac12\mathbf{B}\times\mathbf{r}$. Check for $\mathbf{B} = B\hat{\mathbf{z}}$: $\mathbf{A} = \tfrac12B(-y, x, 0)$, the rotation field of module 1, whose curl is $(0, 0, \tfrac12B + \tfrac12B) = B\hat{\mathbf{z}}$. It is also divergence-free, so it is in the Coulomb gauge. Any gradient could be added: $\mathbf{A} = (-By, 0, 0)$ has the same curl and is sometimes more convenient.

**Worked example 13 (magnetic dipole).** Far from a small loop with moment $\mathbf{m} = I\mathbf{S}$, $\mathbf{A} = \dfrac{\mu_0}{4\pi}\dfrac{\mathbf{m}\times\hat{\mathbf{r}}}{r^2}$. For $\mathbf{m} = m\hat{\mathbf{z}}$ this is $A_\phi = \dfrac{\mu_0 m\sin\theta}{4\pi r^2}$, circling the axis like the current that made it. Its curl is the dipole field of primer 00a. The flux through the coaxial disk of worked example 2 follows in one line: on the disk's rim $r^2 = a^2 + h^2$ and $\sin\theta = a/r$, so $\Phi_B = \oint\mathbf{A}\cdot d\mathbf{l} = A_\phi\cdot2\pi a = \dfrac{\mu_0 m}{2}\dfrac{a^2}{(a^2 + h^2)^{3/2}}$, the same result as the area integral, with no integration.

The code computes $\mathbf{A}$ of a circular loop on a 3D grid by the segment sum $\frac{\mu_0 I}{4\pi}\sum d\mathbf{l}/R$, takes its curl with `fields.curl`, and compares with the Biot-Savart field from `bs.wire_B` at grid points away from the wire. The comparison is done at two grid spacings so that the second-order convergence of the central-difference curl is visible. It then checks worked example 12 on a grid.

```python
def wire_A(mid, dl, I, obs, chunk=128):
    '''Vector potential of current I on segments (mid, dl): (mu0 I / 4 pi) sum of dl / R.'''
    obs = np.asarray(obs, dtype=float)
    flat = obs.reshape(-1, 3)
    A = np.zeros_like(flat)
    for start in range(0, len(mid), chunk):
        m_, d_ = mid[start:start + chunk], dl[start:start + chunk]
        dist = np.linalg.norm(flat[:, None, :] - m_[None, :, :], axis=-1)
        with np.errstate(divide="ignore"):
            A += np.where(dist > 0, 1.0 / dist, 0.0) @ d_
    return MU0 * I / (4 * np.pi) * A.reshape(obs.shape)

I, R = 1.0, 0.5
mid, dl = bs.segments(bs.loop(R, 300))
for n in (17, 33):
    x, y, z, P = grids.grid3d((-1, 1), (-1, 1), (-1, 1), n)
    h3 = grids.spacing(x, y, z)
    A = wire_A(mid, dl, I, P)
    B_from_A = fields.curl(A, *h3)
    B_direct = bs.wire_B(mid, dl, I, P, chunk=128)
    dist_to_wire = np.sqrt((np.hypot(P[..., 0], P[..., 1]) - R) ** 2 + P[..., 2] ** 2)
    keep = dist_to_wire > 0.35
    keep[:2] = keep[-2:] = False; keep[:, :2] = keep[:, -2:] = False; keep[:, :, :2] = keep[:, :, -2:] = False
    rel = np.linalg.norm(B_from_A - B_direct, axis=-1)[keep] / np.linalg.norm(B_direct, axis=-1)[keep]
    divA = np.abs(fields.divergence(A, *h3))[keep] / np.linalg.norm(B_direct, axis=-1)[keep]
    print(f"grid {n}^3, h = {h3[0]:.4f}: |curl A - B| / |B| at points > 0.35 from the wire: median {np.median(rel):.1e}, max {rel.max():.1e};   |div A| / |B|: median {np.median(divA):.1e}")
A_pt = wire_A(mid, dl, I, np.array([0.7, 0.0, 0.1]))
print(f"A points along the current: A at (0.7, 0, 0.1) = {A_pt * 1e7} x 1e-7 T m   (the loop runs along +y there)")

B0 = np.array([0.1, -0.2, 0.3])
A_uniform = 0.5 * np.cross(B0, P)
print("curl of (1/2) B0 x r on the grid, at one point:", fields.curl(A_uniform, *h3)[16, 16, 16], "  B0 =", B0)
```

### 3.6 Why the symmetry method is the exception

A single circular loop has the same rotational symmetry as the solenoid, but no translational symmetry along its axis, and that is enough to break the method. $\mathbf{B}$ depends on both $r$ and $z$, so no circle or rectangle has $|\mathbf{B}|$ constant along it. Ampere's law still holds for every loop you draw, as the code above shows for loops of any shape around a wire, but it produces one number per loop, not a field. The same is true of a finite solenoid, a Helmholtz pair, a bent wire, a square loop. These are the generic case, and for them there are two routes: the Biot-Savart sum, which is what `bs.wire_B` does and what notebook 01 uses for loops and Helmholtz coils; or the three Poisson equations for $\mathbf{A}$ on a grid, solved with the machinery of module 2 and followed by a curl. The symmetric cases of section 3.3 are valuable as tests for both.

## 4. Verify

Ampere's law on the solenoid, with a loop that the symmetry argument of worked example 8 used. The circuit is the 100-turn helix closed by a straight return wire well outside it, so that the current is steady and Ampere's law applies exactly. The Amperian loop is a rectangle in a plane through the axis: one long side on the axis, one outside the winding, two radial sides that cut through the winding. The enclosed current is $I$ times the number of turns that pierce the rectangle, which the code counts from the helix vertices. The two radial sides pass within half a pitch of the wire, where the field is sharp, so they are sampled finely, and the loop is placed midway between turns.

One numerical point is shown rather than hidden. `bs.wire_B` evaluates each segment's contribution at the segment's midpoint, a one-point rule that is accurate only at distances larger than the segment length. Near the winding the radial sides are closer to the wire than one segment, so the circulation misses by a fraction that falls by four each time the segments are halved: second order, like every central difference in this course. The finest helix brings the mismatch below $10^{-4}$. A rectangle entirely outside the winding gives zero, and the off-centre box flux of section 3.1 is checked again here with an assertion.

```python
R_v, L_v, turns_v, I_v = 0.05, 1.0, 100, 2.0
pitch = L_v / turns_v
z1, z2 = -0.2 - pitch / 2, 0.2 + pitch / 2                   # radial sides midway between turns
corners = np.array([[0, 0, z1], [3 * R_v, 0, z1], [3 * R_v, 0, z2], [0, 0, z2]])
n_pts = [2000, 400, 2000, 400]                               # fine sampling on the radial sides
rect = np.vstack([np.linspace(corners[k], corners[(k + 1) % 4], n_pts[k], endpoint=False) for k in range(4)] + [corners[:1]])
print("segments per turn   circulation / (mu0 I)   turns piercing the rectangle")
for per_turn in (60, 120, 240):
    circuit = np.vstack([bs.helix(R_v, L_v, turns_v, per_turn), [[5 * R_v, 0, L_v / 2], [5 * R_v, 0, -L_v / 2]]])
    mid_c, dl_c = bs.segments(circuit, closed=True)
    circ = fields.line_integral(lambda p: bs.wire_B(mid_c, dl_c, I_v, p), rect)
    yv = circuit[:, 1]
    crossing = (yv[:-1] < 0) & (yv[1:] >= 0) & (circuit[:-1, 0] > 0)          # wire passes through the half-plane y = 0, x > 0
    z_cross = 0.5 * (circuit[:-1, 2] + circuit[1:, 2])[crossing]
    enclosed = int(np.sum((z_cross > z1) & (z_cross < z2)))
    print(f"{per_turn:8d}           {circ / (MU0 * I_v):+12.5f}              {enclosed}   (loop runs down the axis, so the sign is negative)")
assert abs(-circ / (MU0 * I_v) / enclosed - 1.0) < 2e-4

outer = np.array([[3 * R_v, 0, z1], [4 * R_v, 0, z1], [4 * R_v, 0, z2], [3 * R_v, 0, z2]])
rect_out = np.vstack([np.linspace(outer[k], outer[(k + 1) % 4], 400, endpoint=False) for k in range(4)] + [outer[:1]])
circ_out = fields.line_integral(lambda p: bs.wire_B(mid_c, dl_c, I_v, p), rect_out)
print(f"\nrectangle entirely outside the winding: circulation / (mu0 I) = {circ_out / (MU0 * I_v):+.1e}")
assert abs(circ_out / (MU0 * I_v)) < 1e-4
assert abs(flux_box) / unsigned < 1e-3
print("Ampere and flux checks passed")
```

## 7. Warm-up answers

1. $\Phi_B = \int_S\mathbf{B}\cdot d\mathbf{S}$, the normal component of $\mathbf{B}$ summed over the surface. Unit: weber, $1\,\mathrm{Wb} = 1\,\mathrm{T\,m^2}$.
2. Zero. Field lines of $\mathbf{B}$ have no beginning or end (no magnetic charge), so every line entering a closed surface also leaves it. Locally, $\nabla\cdot\mathbf{B} = 0$.
3. $\oint_C\mathbf{B}\cdot d\mathbf{l} = \mu_0 I_{\text{enc}}$, with $I_{\text{enc}}$ the net current through any surface bounded by $C$. Right-hand rule: fingers along the direction of travel around $C$, thumb gives the direction in which current counts positive.
4. $\nabla\cdot\mathbf{B} = 0$ and $\nabla\times\mathbf{B} = \mu_0\mathbf{J}$, beside $\nabla\cdot\mathbf{E} = \rho/\epsilon_0$ and $\nabla\times\mathbf{E} = 0$.
5. Symmetry makes $\mathbf{B}$ azimuthal with magnitude depending on $r$ only; on a circle of radius $r$, $B\cdot2\pi r = \mu_0 I$, so $B = \mu_0 I/2\pi r$.
6. A rectangular loop with both long sides outside encloses no current, so the outside field is the same at every radius, hence zero (it vanishes far away). A loop with one side inside encloses $n\ell I$ and gives $B\ell = \mu_0 n\ell I$ regardless of where the inside side sits, so the inside field is uniform, $\mu_0 nI$.
7. $\nabla\cdot\mathbf{B} = 0$ and the divergence of any curl is zero, so $\mathbf{B} = \nabla\times\mathbf{A}$ is always possible. The electrostatic $\mathbf{E}$ has non-zero divergence wherever there is charge, so it cannot be a pure curl; instead it has zero curl, which is why it is a gradient.
