## 2. The ideas

**A gap in Ampere's law.** In module 4, Ampere's law related the circulation of $\mathbf{B}$ around a loop to the current through a surface bounded by that loop, and it did not matter which surface: for steady currents every surface collects the same current, because $\nabla\cdot\mathbf{J} = 0$ means no current can start or stop. A capacitor that is being charged breaks this. Current flows along the wire into one plate and out of the other, but none crosses the gap. Take a loop around the wire. A flat surface spanning the loop is pierced by the wire and collects the current $I$. A bag-shaped surface with the same rim that passes between the plates collects nothing. The same circulation cannot equal both $\mu_0 I$ and zero. The flaw is the assumption $\nabla\cdot\mathbf{J} = 0$: charge is piling up on the plates, and where charge accumulates the current density has a divergence.

**Maxwell's fix.** Charge that piles up on a plate also builds an electric field in the gap. Gauss' law ties the two together: the rate at which charge arrives equals $\epsilon_0$ times the rate at which the electric flux grows. So the quantity $\epsilon_0\,\partial\mathbf{E}/\partial t$ takes over exactly where the conduction current stops. Maxwell added it to Ampere's law as a second source of magnetic field, the displacement current. With it, every surface bounded by the loop gives the same answer, and the law is consistent with conservation of charge. The new term is tiny in a circuit, but it is the term that makes light possible.

**The complete theory.** The corrected set has four equations. Gauss' law says charge is the source of $\mathbf{E}$. The second says $\mathbf{B}$ has no sources: its field lines never end. Faraday's law says a changing $\mathbf{B}$ makes $\mathbf{E}$ circulate. The Ampere-Maxwell law says currents and a changing $\mathbf{E}$ make $\mathbf{B}$ circulate. Together with the Lorentz force $\mathbf{F} = q(\mathbf{E} + \mathbf{v}\times\mathbf{B})$, which says how the fields push on charges, these equations are the whole of classical electromagnetism. Every static result of modules 1 to 4 is a special case in which the time derivatives vanish.

**Fields that sustain each other.** The two curl equations form a loop. A changing $\mathbf{E}$ produces a circulating $\mathbf{B}$, and a changing $\mathbf{B}$ produces a circulating $\mathbf{E}$. Far from all charges and currents nothing else is needed: a disturbance in one field drives the other, which drives the first again, and the pattern moves. Combining the two curl equations eliminates one field and leaves the wave equation for the other, a law that says any shape moves without change at one fixed speed.

**The speed of light from two constants.** The speed in that wave equation is $1/\sqrt{\mu_0\epsilon_0}$. The constant $\epsilon_0$ comes from measuring forces between charges at rest, and $\mu_0$ from forces between currents. Neither experiment involves light, yet the combination is $2.998\times10^8$ m/s, the measured speed of light. Maxwell drew the conclusion that light is an electromagnetic wave, and every later test has agreed: radio waves, microwaves, infrared, visible light, X-rays and gamma rays are the same thing at different frequencies.

**Energy on the move.** A wave carries energy. The energy density of the fields, $\tfrac12\epsilon_0E^2$ for the electric part and $B^2/2\mu_0$ for the magnetic part, met in modules 2 and 6, moves with the wave. Maxwell's equations give a bookkeeping law for it, Poynting's theorem: energy in a region drops only by flowing out through its surface or by doing work on charges. The flow is the Poynting vector $\mathbf{S} = \mathbf{E}\times\mathbf{B}/\mu_0$, in watts per square metre. Sunlight, a laser beam and a radio signal are all values of $\mathbf{S}$.

**Waves in matter.** In a dielectric the polarization of module 3 slows the wave to $c/n$ with $n = \sqrt{\epsilon_r}$. Where the medium changes, the boundary conditions on the fields force part of the wave back: the reflection and transmission coefficients follow in three lines from continuity of the tangential fields. In a conductor the conduction current swamps the displacement current, the wave equation turns into a diffusion equation, and the field dies within a skin depth.

### 3.1 The displacement current (section 17.1)

Conservation of charge is local: charge leaving a small volume must cross its surface. In differential form (primer M5.0a, section 3.5),
$$\nabla\cdot\mathbf{J} + \frac{\partial\rho}{\partial t} = 0.$$
Steady currents have $\partial\rho/\partial t = 0$ and so $\nabla\cdot\mathbf{J} = 0$, which is what Ampere's law $\nabla\times\mathbf{B} = \mu_0\mathbf{J}$ silently assumes: the divergence of a curl is always zero (primer M1.0c, section 3.6), so the left side has zero divergence and the right side must too. When charge accumulates the right side has a divergence and the equation is contradictory.

The repair uses Gauss' law, $\rho = \epsilon_0\nabla\cdot\mathbf{E}$. Substitute it into the continuity equation:
$$\nabla\cdot\mathbf{J} + \epsilon_0\frac{\partial}{\partial t}\nabla\cdot\mathbf{E} = \nabla\cdot\left(\mathbf{J} + \epsilon_0\frac{\partial\mathbf{E}}{\partial t}\right) = 0.$$
The combination in brackets always has zero divergence, whatever the charges do. Maxwell replaced $\mathbf{J}$ in Ampere's law by this combination:
$$\nabla\times\mathbf{B} = \mu_0\mathbf{J} + \mu_0\epsilon_0\frac{\partial\mathbf{E}}{\partial t}, \qquad \oint\mathbf{B}\cdot d\mathbf{l} = \mu_0 I_{\text{enc}} + \mu_0\epsilon_0\frac{d\Phi_E}{dt},$$
where $\Phi_E = \int\mathbf{E}\cdot d\mathbf{S}$ is the electric flux through the surface. The term $\epsilon_0\,\partial\mathbf{E}/\partial t$ has units of current density and is called the displacement current density $\mathbf{J}_d$. The name is historical: it is not a flow of charge. It is a changing electric field, and it makes magnetic field exactly as a real current of the same size would.

Now the capacitor paradox resolves. With charge $Q$ on plates of area $A$, the field in the gap is $E = Q/\epsilon_0A$ (primer M2.0a, worked example 12), so the displacement current through the gap is $\epsilon_0A\,dE/dt = dQ/dt = I$. The flat surface collects the conduction current $I$, the bag between the plates collects the displacement current $I$, and the circulation is $\mu_0 I$ either way.

**Worked example 1 (the size of the displacement current).** Circular plates of radius $a = 5\,$cm are charged with a current $I = 2\,$A. The area is $\pi a^2 = 7.854\times10^{-3}\,\mathrm{m^2}$, so the field in the gap grows at
$$\frac{dE}{dt} = \frac{I}{\epsilon_0\pi a^2} = \frac{2}{8.854\times10^{-12}\times7.854\times10^{-3}} = 2.876\times10^{13}\ \mathrm{V/(m\,s)}.$$
The displacement current density is $\epsilon_0\,dE/dt = I/\pi a^2 = 254.6\,\mathrm{A/m^2}$, the same as the conduction current density would be if the current filled a wire of radius $a$. An enormous rate of change of $E$ is needed to match a modest current: that is why the term went unnoticed in every static and slowly varying experiment.

**Worked example 2 (the magnetic field between the plates).** Take a circle of radius $r < a$ between the plates, coaxial with them. No conduction current crosses it, but the displacement current through it is the fraction $r^2/a^2$ of $I$. By symmetry $\mathbf{B}$ circles the axis with constant magnitude, so
$$B\cdot2\pi r = \mu_0\epsilon_0\,\pi r^2\frac{dE}{dt} \quad\Longrightarrow\quad B = \frac{\mu_0\epsilon_0r}{2}\frac{dE}{dt} = \frac{\mu_0 I r}{2\pi a^2}.$$
At $r = 2.5\,$cm: $B = 4\pi\times10^{-7}\times2\times0.025/(2\pi\times0.0025) = 4.0\times10^{-6}\,$T. At the rim, $r = a$, it reaches $\mu_0I/2\pi a = 8.0\times10^{-6}\,$T, and outside the plates it falls as $\mu_0 I/2\pi r$, exactly the field around the wire. The gap looks, magnetically, like a thick wire carrying the current $I$ uniformly (compare the thick wire of primer M4.0b, worked example 6).

The code integrates $\epsilon_0\,\partial E/\partial t$ over disks of growing radius in the gap, with the field uniform inside the plates and zero outside (fringes ignored), and compares the circulation it predicts with the one from the wire.

```python
a, I = 0.05, 2.0                                   # plate radius (m), charging current (A)
dEdt = I / (EPS0 * np.pi * a**2)
print(f"dE/dt in the gap = {dEdt:.3e} V/(m s);   eps0 dE/dt = {EPS0 * dEdt:.1f} A/m^2 = I/(pi a^2) = {I / (np.pi * a**2):.1f} A/m^2")

def displacement_current(r, nr=2000):
    '''eps0 dE/dt integrated over a disk of radius r between the plates (midpoint rule in s).

    The field is uniform inside the plates' radius a and zero outside, so the integral runs to min(r, a).
    '''
    edge = min(r, a)
    s = (np.arange(nr) + 0.5) * edge / nr
    return np.sum(EPS0 * dEdt * 2 * np.pi * s * (edge / nr))

print("\n   r [cm]   I_d through the gap disk   circulation / mu0 via the gap   via the wire (flat disk)")
for r in (0.01, 0.025, 0.05, 0.08, 0.15):
    Id = displacement_current(r)
    print(f"   {100 * r:5.1f}    {Id:10.4f} A               {Id:10.4f} A                     {I:.4f} A" + ("   (circle smaller than the plates)" if r < a else ""))

r_line = np.linspace(0.002, 0.15, 300)
B_gap = np.array([MU0 * displacement_current(r) / (2 * np.pi * r) for r in r_line])
plt.plot(100 * r_line, 1e6 * B_gap, label="between the plates: displacement current")
plt.plot(100 * r_line, 1e6 * MU0 * I / (2 * np.pi * r_line), "--", label="around the wire: mu0 I / 2 pi r")
plt.axvline(100 * a, color="grey", ls=":"); plt.ylim(0, 25)
plt.xlabel("r [cm]"); plt.ylabel("B [micro T]"); plt.legend(); plt.title("magnetic field of a charging capacitor"); plt.show()
print(f"B at r = 2.5 cm: {1e6 * MU0 * displacement_current(0.025) / (2 * np.pi * 0.025):.3f} micro T;  at r = a: {1e6 * MU0 * displacement_current(a) / (2 * np.pi * a):.3f} micro T")
```

### 3.2 Maxwell's equations (section 17.2)

With the displacement current in place, the four equations read:

| law | integral form | differential form | in words |
|---|---|---|---|
| Gauss | $\oint\mathbf{E}\cdot d\mathbf{S} = Q_{\text{enc}}/\epsilon_0$ | $\nabla\cdot\mathbf{E} = \rho/\epsilon_0$ | charge is the source of $\mathbf{E}$ |
| no magnetic charge | $\oint\mathbf{B}\cdot d\mathbf{S} = 0$ | $\nabla\cdot\mathbf{B} = 0$ | $\mathbf{B}$ field lines never end |
| Faraday | $\oint\mathbf{E}\cdot d\mathbf{l} = -\dfrac{d\Phi_B}{dt}$ | $\nabla\times\mathbf{E} = -\dfrac{\partial\mathbf{B}}{\partial t}$ | a changing $\mathbf{B}$ makes $\mathbf{E}$ circulate |
| Ampere-Maxwell | $\oint\mathbf{B}\cdot d\mathbf{l} = \mu_0I_{\text{enc}} + \mu_0\epsilon_0\dfrac{d\Phi_E}{dt}$ | $\nabla\times\mathbf{B} = \mu_0\mathbf{J} + \mu_0\epsilon_0\dfrac{\partial\mathbf{E}}{\partial t}$ | currents and a changing $\mathbf{E}$ make $\mathbf{B}$ circulate |

The integral and differential forms are the same statements, connected by the divergence theorem for the first two and Stokes' theorem for the last two (primer M1.0d). The sources are the charge density $\rho$ in C/m³ and the current density $\mathbf{J}$ in A/m². The constants are $\epsilon_0 = 8.854\times10^{-12}$ F/m and $\mu_0 = 1.2566\times10^{-6}$ H/m. To say what the fields do, one more law is needed, the Lorentz force on a charge $q$ moving with velocity $\mathbf{v}$:
$$\mathbf{F} = q(\mathbf{E} + \mathbf{v}\times\mathbf{B}).$$

**The static limit.** When nothing changes in time, the two curl equations lose their time derivatives. Then $\nabla\times\mathbf{E} = 0$, so $\mathbf{E} = -\nabla V$ (module 1), and Gauss' law becomes Poisson's equation (module 2). The magnetic pair becomes $\nabla\cdot\mathbf{B} = 0$ and $\nabla\times\mathbf{B} = \mu_0\mathbf{J}$, the magnetostatics of module 4, solved by Biot-Savart. Electric and magnetic fields decouple and can be studied separately, which is what modules 1 to 4 did. Faraday's law (primer 00a) was the first coupling: a changing $\mathbf{B}$ makes $\mathbf{E}$. The displacement current is the second: a changing $\mathbf{E}$ makes $\mathbf{B}$.

**Worked example 3 (charge conservation is built in).** Take the divergence of the Ampere-Maxwell law. The left side is the divergence of a curl, which is zero. The right side gives
$$0 = \mu_0\nabla\cdot\mathbf{J} + \mu_0\epsilon_0\frac{\partial}{\partial t}\nabla\cdot\mathbf{E} = \mu_0\left(\nabla\cdot\mathbf{J} + \frac{\partial\rho}{\partial t}\right),$$
using Gauss' law in the last step. The continuity equation is not an extra assumption. It follows from the field equations, and without the displacement current it would not: the original Ampere law would force $\nabla\cdot\mathbf{J} = 0$ always, which is false for any capacitor.

**The asymmetry.** In empty space ($\rho = 0$, $\mathbf{J} = 0$) the equations are almost symmetric between $\mathbf{E}$ and $\mathbf{B}$: swapping $\mathbf{E}\to c\mathbf{B}$ and $c\mathbf{B}\to-\mathbf{E}$ leaves them unchanged. With sources the symmetry breaks, because there are electric charges and currents but no magnetic charges. No magnetic monopole has ever been found. The equation $\nabla\cdot\mathbf{B} = 0$ records that fact, and the missing source terms in Faraday's law are the same fact seen from the other side.

### 3.3 The wave equation

Go to empty space: no charge and no current. The equations become
$$\nabla\cdot\mathbf{E} = 0, \qquad \nabla\cdot\mathbf{B} = 0, \qquad \nabla\times\mathbf{E} = -\frac{\partial\mathbf{B}}{\partial t}, \qquad \nabla\times\mathbf{B} = \mu_0\epsilon_0\frac{\partial\mathbf{E}}{\partial t}.$$
Take the curl of Faraday's law. On the right, the curl and the time derivative commute, and the Ampere-Maxwell law replaces $\nabla\times\mathbf{B}$:
$$\nabla\times(\nabla\times\mathbf{E}) = -\frac{\partial}{\partial t}(\nabla\times\mathbf{B}) = -\mu_0\epsilon_0\frac{\partial^2\mathbf{E}}{\partial t^2}.$$
On the left, use the identity $\nabla\times(\nabla\times\mathbf{F}) = \nabla(\nabla\cdot\mathbf{F}) - \nabla^2\mathbf{F}$ from `study/vector_calculus_ref.md`, where $\nabla^2$ acts on each Cartesian component. Gauss' law in vacuum removes the first term. What remains is
$$\nabla^2\mathbf{E} = \mu_0\epsilon_0\frac{\partial^2\mathbf{E}}{\partial t^2} = \frac{1}{c^2}\frac{\partial^2\mathbf{E}}{\partial t^2}, \qquad c = \frac{1}{\sqrt{\mu_0\epsilon_0}}.$$
Starting from the curl of the Ampere-Maxwell law instead gives the same equation for $\mathbf{B}$. Each Cartesian component of each field obeys the wave equation with speed $c$.

**Worked example 4 (the speed).** $\mu_0\epsilon_0 = 1.25664\times10^{-6}\times8.85419\times10^{-12} = 1.11265\times10^{-17}\,\mathrm{s^2/m^2}$, and $c = 1/\sqrt{1.11265\times10^{-17}} = 2.99792\times10^8\,$m/s. Since 1983 the metre has been defined so that $c = 299\,792\,458$ m/s exactly. Since 2019, $\mu_0$ is a measured quantity rather than exactly $4\pi\times10^{-7}$. The library uses the older exact value, which differs from the measured one by about five parts in $10^{10}$, so the code reproduces $c$ to about $3\times10^{-10}$.

**What the wave equation says.** In one dimension, with the field depending on $x$ and $t$ only,
$$\frac{\partial^2 u}{\partial x^2} = \frac{1}{c^2}\frac{\partial^2 u}{\partial t^2}.$$
Any function of the combination $x - ct$ solves it. Let $s = x - ct$ and $u = f(s)$. Then $\partial u/\partial x = f'(s)$ and $\partial^2u/\partial x^2 = f''(s)$, while $\partial u/\partial t = -c f'(s)$ and $\partial^2u/\partial t^2 = c^2 f''(s)$. The two sides agree for any $f$. The same works for $g(x + ct)$, and because the equation is linear the sum $f(x - ct) + g(x + ct)$ is the general solution (d'Alembert's). The meaning: whatever shape $f$ has at $t = 0$, at time $t$ the same shape sits a distance $ct$ further along $+x$. The function $g$ moves the other way. Nothing in the equation favours a particular shape, frequency or direction: pulses, sine waves and anything else all move at $c$ without changing form.

**Worked example 5 (a pulse in flight).** A pulse $f(x) = E_0\exp(-x^2/w^2)$ is centred at $x = 0$ at $t = 0$ and moves along $+x$. At $t = 10\,$ns its centre is at $ct = 2.998\,$m, with exactly the same shape. Light covers 30 cm in one nanosecond, which is why a 3 GHz processor clock cycle (0.33 ns) lasts as long as light takes to cross 10 cm, and why primer M5.0c, section 3.8, put the limit of the lumped circuit model at sizes comparable with the wavelength.

The code checks d'Alembert's solution with SymPy and then checks a sinusoidal wave against all four of Maxwell's vacuum equations, along with two fields that look plausible and fail.

```python
x, y, z, t = sp.symbols("x y z t", real=True)
mu0_s, eps0_s, E0_s, k_s = sp.symbols("mu_0 epsilon_0 E_0 k", positive=True)
c_s = 1 / sp.sqrt(mu0_s * eps0_s)
f, g = sp.Function("f"), sp.Function("g")

u = f(x - c_s * t) + g(x + c_s * t)
print("d'Alembert: d2u/dt2 - c^2 d2u/dx2 =", sp.simplify(sp.diff(u, t, 2) - c_s**2 * sp.diff(u, x, 2)))

def div(F):
    return sum(sp.diff(Fi, v) for Fi, v in zip(F, (x, y, z)))

def curl(F):
    return [sp.diff(F[2], y) - sp.diff(F[1], z), sp.diff(F[0], z) - sp.diff(F[2], x), sp.diff(F[1], x) - sp.diff(F[0], y)]

def maxwell_residuals(E, B):
    '''Left minus right side of the four vacuum equations; all zero for a solution.'''
    res = [div(E), div(B)]
    res += [cE + sp.diff(Bi, t) for cE, Bi in zip(curl(E), B)]                    # Faraday
    res += [cB - mu0_s * eps0_s * sp.diff(Ei, t) for cB, Ei in zip(curl(B), E)]    # Ampere-Maxwell
    return [sp.simplify(r) for r in res]

phase = k_s * x - c_s * k_s * t                       # omega = c k
E_pw = [0, E0_s * sp.cos(phase), 0]
B_pw = [0, 0, E0_s / c_s * sp.cos(phase)]
print("plane wave, B = E/c along z:     ", maxwell_residuals(E_pw, B_pw))
print("same E, B = E (wrong size):       ", [r for r in maxwell_residuals(E_pw, [0, 0, E0_s * sp.cos(phase)]) if r != 0][:1], "... fails")
print("longitudinal E along x:           gauss residual =", maxwell_residuals([E0_s * sp.cos(phase), 0, 0], [0, 0, 0])[0])
print(f"\nc = 1/sqrt(mu0 eps0) = {C:.6e} m/s   (defined value 299792458; relative difference {C / 299792458 - 1:.1e})")
print(f"a pulse travels {C * 10e-9:.3f} m in 10 ns and {100 * C * 1e-9:.2f} cm in 1 ns")
```

### 3.4 Plane waves

The simplest solution is a plane wave: fields that depend only on $x$ and $t$, so they are the same everywhere on each plane $x = $ constant. Take a sinusoidal electric field along $y$,
$$\mathbf{E} = E_0\cos(kx - \omega t)\,\hat{\mathbf{y}}.$$
It solves the wave equation if $\omega = ck$. Here $k$ is the wavenumber, the wavelength is $\lambda = 2\pi/k$, the frequency is $f = \omega/2\pi$, and the relation $\omega = ck$ is the same as $\lambda f = c$.

**The magnetic field follows.** Faraday's law fixes $\mathbf{B}$. The only non-zero component of $\nabla\times\mathbf{E}$ is the $z$ component, $\partial E_y/\partial x = -kE_0\sin(kx - \omega t)$, so
$$\frac{\partial B_z}{\partial t} = -\frac{\partial E_y}{\partial x} = kE_0\sin(kx - \omega t) \quad\Longrightarrow\quad B_z = \frac{k}{\omega}E_0\cos(kx - \omega t) = \frac{E_0}{c}\cos(kx - \omega t).$$
(An added constant would be a static field, not part of the wave.) So $\mathbf{B}$ points along $z$, is in phase with $\mathbf{E}$, and is smaller by the factor $c$: $B = E/c$. The SymPy check above confirms that this pair satisfies all four equations, and that $\mathbf{B} = \mathbf{E}$ in size does not.

**Transversality.** Could the wave have a component along its direction of travel? An $E_x$ that depends on $x$ has $\nabla\cdot\mathbf{E} = \partial E_x/\partial x \neq 0$, which Gauss' law forbids in empty space; the SymPy check shows that residual. The same argument with $\nabla\cdot\mathbf{B} = 0$ rules out a longitudinal $B_x$. So electromagnetic waves are transverse: $\mathbf{E}$, $\mathbf{B}$ and the direction of travel $\hat{\mathbf{k}}$ are mutually perpendicular, ordered so that $\mathbf{E}\times\mathbf{B}$ points along $\hat{\mathbf{k}}$. In vector form, $\mathbf{B} = \hat{\mathbf{k}}\times\mathbf{E}/c$.

**Polarisation.** The direction of $\mathbf{E}$ is the polarisation. The wave above is linearly polarised along $y$. A wave with $E_y = E_0\cos(kx - \omega t)$ and $E_z = E_0\sin(kx - \omega t)$ has an $\mathbf{E}$ of constant length that rotates as the wave passes: circular polarisation. Any polarisation is a sum of two linear ones at right angles with some relative phase. Polarising sunglasses pass one linear component, and 3D cinema glasses separate the two circular ones.

**Worked example 6 (field sizes and scales).** An FM radio wave at $f = 100\,$MHz with $E_0 = 100\,$mV/m, a strong signal near a transmitter. Wavelength $\lambda = c/f = 2.998\,$m, wavenumber $k = 2\pi/\lambda = 2.096\,$rad/m, angular frequency $\omega = 6.283\times10^8\,$rad/s. The magnetic amplitude is $B_0 = E_0/c = 0.1/2.998\times10^8 = 3.3\times10^{-10}\,$T, a hundred thousand times smaller than Earth's field. Both fields carry the same energy, as section 3.5 shows; the small number reflects the units, not a weak magnetic part.

**Worked example 7 (the spectrum).** The same equations hold at every frequency; only $\lambda = c/f$ changes.

### 3.5 Energy and the Poynting vector

The energy density stored in electric and magnetic fields is
$$u = \frac{\epsilon_0E^2}{2} + \frac{B^2}{2\mu_0}.$$
The electric part was derived for a capacitor in module 2 (primer M2.0c, section 3.6), the magnetic part for an inductor in primer 00b of this module. Both were shown to hold point by point, not just for those devices.

**Worked example 8 (equal shares).** In a plane wave $B = E/c$, so the magnetic part is $B^2/2\mu_0 = E^2/(2\mu_0c^2) = \epsilon_0E^2/2$, using $\mu_0c^2 = 1/\epsilon_0$. The magnetic field carries exactly as much energy as the electric field, at every point and every instant, even though its value in tesla looks tiny. The total is $u = \epsilon_0E^2$.

**Poynting's theorem.** Energy conservation for fields follows from Maxwell's equations. Start from the vector identity
$$\nabla\cdot(\mathbf{E}\times\mathbf{B}) = \mathbf{B}\cdot(\nabla\times\mathbf{E}) - \mathbf{E}\cdot(\nabla\times\mathbf{B}).$$
Faraday's law gives $\mathbf{B}\cdot(\nabla\times\mathbf{E}) = -\mathbf{B}\cdot\partial\mathbf{B}/\partial t = -\partial(B^2/2)/\partial t$. Ampere-Maxwell gives $\mathbf{E}\cdot(\nabla\times\mathbf{B}) = \mu_0\mathbf{J}\cdot\mathbf{E} + \mu_0\epsilon_0\,\partial(E^2/2)/\partial t$. Substitute and divide by $\mu_0$:
$$\frac{\partial u}{\partial t} + \nabla\cdot\mathbf{S} = -\mathbf{J}\cdot\mathbf{E}, \qquad \mathbf{S} = \frac{1}{\mu_0}\mathbf{E}\times\mathbf{B} = \mathbf{E}\times\mathbf{H}.$$
Integrate over a volume and use the divergence theorem. The field energy inside decreases at the rate it flows out through the surface, $\oint\mathbf{S}\cdot d\mathbf{S}$, plus the rate it is handed to charges, $\int\mathbf{J}\cdot\mathbf{E}\,dv$. That last term is the Joule heating of primer M5.0a, section 3.4. So $\mathbf{S}$, the Poynting vector, is the energy flux: watts per square metre crossing a surface perpendicular to it. Here $\mathbf{H} = \mathbf{B}/\mu_0$ in vacuum (primer M4.0c, section 3.4).

**The plane wave's energy flow.** With $\mathbf{E}$ along $y$ and $\mathbf{B}$ along $z$, $\mathbf{E}\times\mathbf{B}$ points along $x$, the direction of travel, with magnitude $S = EB/\mu_0 = E^2/(\mu_0c)$. Two consistency checks. First, $S = uc$: the energy density moves at the speed $c$. Second, the combination $\mu_0c = \sqrt{\mu_0/\epsilon_0} = \eta_0 = 376.73\,\Omega$ is the impedance of free space, so $S = E^2/\eta_0$ and $H = E/\eta_0$: the vacuum acts like a resistance relating $E$ to $H$. The time average of $\cos^2$ is $1/2$, so the intensity, the average of $S$, is
$$I = \langle S\rangle = \frac{E_0^2}{2\eta_0} = \frac{c\,\epsilon_0E_0^2}{2}.$$

**Worked example 9 (sunlight).** Above the atmosphere the Sun delivers $I = 1361\,\mathrm{W/m^2}$. The field amplitude is $E_0 = \sqrt{2\eta_0I} = \sqrt{2\times376.73\times1361} = 1013\,$V/m, and $B_0 = E_0/c = 3.38\,\mu$T. Absorbed light also pushes: the momentum flux is $I/c$, a radiation pressure of $4.5\times10^{-6}\,$Pa, which is what a solar sail uses.

**Worked example 10 (a laser pointer).** A 1 mW beam in a spot of 1 mm diameter has $I = 10^{-3}/(\pi\times(0.5\times10^{-3})^2) = 1273\,\mathrm{W/m^2}$, close to full sunlight, with $E_0 = 980\,$V/m. The eye focuses that parallel beam to a spot a few micrometres across, raising the intensity by a factor of tens of thousands. That is why a pointer that feels harmless on the skin is dangerous to look into.

The code checks the energy bookkeeping for a Gaussian pulse: the field energy per unit area, $\int u\,dx$, must equal the energy that crosses a fixed plane, $\int S\,dt$, as the pulse passes. Both equal $\epsilon_0E_0^2w\sqrt{\pi/2}$ for a pulse $E_0e^{-(x-ct)^2/w^2}$.

```python
eta0 = np.sqrt(MU0 / EPS0)
print(f"eta0 = sqrt(mu0/eps0) = {eta0:.2f} ohm = mu0 c = {MU0 * C:.2f} ohm")
for name, I_ in (("sunlight", 1361.0), ("laser pointer, 1 mW in 1 mm", 1e-3 / (np.pi * 0.5e-3**2))):
    E0 = np.sqrt(2 * eta0 * I_)
    print(f"{name:28s}: I = {I_:7.1f} W/m^2,  E0 = {E0:6.1f} V/m,  B0 = {E0 / C * 1e6:.3f} micro T,  pressure if absorbed = {I_ / C:.2e} Pa")

# A Gaussian pulse E0 exp(-((x - x0 - c t)/w)^2), B = E/c, moving along +x.
E0p, w, x0, x_probe = 1.0, 0.1, -1.0, 0.5
xg = np.linspace(-3, 3, 60001); dxg = xg[1] - xg[0]
Eg = E0p * np.exp(-((xg - x0) / w) ** 2); Bg = Eg / C
U_E, U_B = np.sum(0.5 * EPS0 * Eg**2) * dxg, np.sum(0.5 * Bg**2 / MU0) * dxg
tg = np.linspace(0, 3.0 / C, 60001)
Ep = E0p * np.exp(-((x_probe - x0 - C * tg) / w) ** 2)
S = Ep * (Ep / C) / MU0                             # E B / mu0 at the probe plane
passed = np.sum(S) * (tg[1] - tg[0])
print(f"\npulse: electric energy {U_E:.5e}, magnetic energy {U_B:.5e} J/m^2 (equal shares)")
print(f"field energy per area {U_E + U_B:.5e} J/m^2;  energy through the plane x = {x_probe} m: {passed:.5e} J/m^2;  formula eps0 E0^2 w sqrt(pi/2) = {EPS0 * E0p**2 * w * np.sqrt(np.pi / 2):.5e}")
```

### 3.6 Waves in matter

**A linear dielectric.** In a linear dielectric without free charge (primer M3.0a), polarization adds bound charge, and its effect is to replace $\epsilon_0$ by $\epsilon_r\epsilon_0$ in Gauss' law and in the displacement current. Repeating section 3.3 gives a wave equation with speed
$$v = \frac{1}{\sqrt{\mu_0\epsilon_r\epsilon_0}} = \frac{c}{n}, \qquad n = \sqrt{\epsilon_r},$$
for a non-magnetic material. The number $n$ is the refractive index. In the medium $B = nE/c$, and the impedance is $\eta = \eta_0/n$, so $H = E/\eta = nE/\eta_0$ and the intensity is $I = nE_0^2/2\eta_0$. The frequency of a wave does not change when it enters a medium, so its wavelength shrinks to $\lambda/n$.

A caveat: $\epsilon_r$ depends on frequency. Water has $\epsilon_r \approx 80$ for static fields because its molecules turn to follow the field (module 3), but at optical frequencies they cannot turn fast enough and only the electron clouds respond, giving $n = 1.33$ rather than $\sqrt{80} \approx 8.9$.

**Reflection and transmission at normal incidence.** A wave travelling along $+x$ in medium 1 meets medium 2 at $x = 0$. Part goes on, part comes back. Write the electric amplitudes at the interface as $E_i$ (incident), $E_r$ (reflected) and $E_t$ (transmitted), all along $y$. For a wave moving along $+x$ the magnetic field is $H_z = +E_y/\eta$. For a wave moving along $-x$ the Poynting vector must point backwards, so $H_z = -E_y/\eta$. Two boundary conditions apply, both from module 3 and 4 arguments with small loops: the tangential $\mathbf{E}$ is continuous (Faraday's law around a thin loop straddling the surface), and the tangential $\mathbf{H}$ is continuous when no free surface current flows (Ampere-Maxwell around the same loop). Here both fields are tangential:
$$E_i + E_r = E_t, \qquad \frac{E_i - E_r}{\eta_1} = \frac{E_t}{\eta_2} \;\Longleftrightarrow\; n_1(E_i - E_r) = n_2E_t.$$
Solving the pair gives the amplitude coefficients
$$r = \frac{E_r}{E_i} = \frac{n_1 - n_2}{n_1 + n_2}, \qquad t = \frac{E_t}{E_i} = \frac{2n_1}{n_1 + n_2}.$$
Intensities carry a factor $n$, so the fractions of power reflected and transmitted are
$$R = r^2, \qquad T = \frac{n_2}{n_1}t^2, \qquad R + T = \frac{(n_1 - n_2)^2 + 4n_1n_2}{(n_1 + n_2)^2} = 1.$$
Energy is conserved, as it must be. The sign of $r$ matters: going into a denser medium ($n_2 > n_1$) $r$ is negative, so the reflected electric field is flipped. Going into a less dense medium $t$ exceeds 1. That is not a gain of energy: the transmitted wave has a larger $E$ but a smaller $n$, and $T$ is still less than 1.

**Worked example 11 (glass and water).** Air to glass, $n_2 = 1.5$: $r = -0.5/2.5 = -0.2$, $R = 4\%$, $t = 0.8$, $T = 1.5\times0.64 = 0.96$. A window has two surfaces and reflects about 8 percent, which is why you see yourself in a window at night. Air to water, $n_2 = 1.33$: $r = -0.142$, $R = 2.0\%$. For the dielectric step of notebook 03, $\epsilon_r = 4$ and $n_2 = 2$: $r = -1/3$, $t = 2/3$, $R = 1/9$ and $T = 8/9$.

**Conductors and the skin depth.** In a conductor Ohm's law adds $\mathbf{J} = \sigma\mathbf{E}$ to the Ampere-Maxwell law. For a sinusoidal field the conduction term $\sigma E$ beats the displacement term $\epsilon_0\omega E$ whenever $\sigma \gg \epsilon_0\omega$; for copper $\sigma/\epsilon_0 = 6.7\times10^{18}\,\mathrm{s^{-1}}$, so that holds far into the infrared. Drop the displacement term and repeat the curl-of-curl step:
$$\nabla^2\mathbf{E} = \mu_0\sigma\frac{\partial\mathbf{E}}{\partial t}.$$
This is a diffusion equation, not a wave equation. Trying $E = E_0e^{i(kx - \omega t)}$ gives $k^2 = i\omega\mu_0\sigma$, so $k = (1 + i)/\delta$ with
$$\delta = \sqrt{\frac{2}{\omega\mu_0\sigma}},$$
and the field varies as $e^{-x/\delta}\cos(x/\delta - \omega t)$: it oscillates and dies within a few skin depths of the surface.

**Worked example 12 (skin depths).** Copper, $\sigma = 5.96\times10^7\,$S/m. At 50 Hz, $\omega\mu_0\sigma = 314.2\times1.2566\times10^{-6}\times5.96\times10^7 = 2.353\times10^4\,\mathrm{m^{-2}}$ and $\delta = \sqrt{2/2.353\times10^4} = 9.2\,$mm: thick power cables gain little from copper deeper than about a centimetre. At 1 GHz, $\delta = 2.1\,\mu$m, so microwave currents flow in a thin skin and a silver plating a few micrometres thick does all the work. Seawater, $\sigma \approx 4\,$S/m, at 10 kHz gives $\delta = 2.5\,$m, and at 1 MHz only 25 cm: submarines can receive only very low frequency radio.

The code runs the 1D solver of notebook 03, `fd.fdtd_1d`, for a pulse hitting a dielectric step with $n_2 = 2$ and with $n_2 = 1.5$, and measures $r$ and $t$ from probes before and after the interface. The pulse is 60 time steps wide so that it stays smooth in the slower medium (primer 00d explains why narrow pulses distort on a grid).

```python
def step_run(eps2, n=2000, dx=1e-3, S=0.5, width_steps=60, i_int=1400, n_steps=5000):
    '''Pulse from a soft source at node 500 hits a step to eps_r = eps2 at node i_int. Returns (r, t, result).'''
    dt = S * dx / fd.C0; w = width_steps * dt; t0 = 4 * w
    eps = np.ones(n + 1); eps[i_int:] = eps2
    res = fd.fdtd_1d(n, dx, n_steps, courant=S, eps_r=eps, source=lambda tt: fd.gaussian(tt, t0, w), src_index=500,
                     probes=[i_int - 100, i_int + 300], snapshot_every=500)
    tt, before, after = res["t"], res["probe_E"][:, 0], res["probe_E"][:, 1]
    inc = before.max()
    late = tt > tt[np.argmax(before)] + 100 * dx / fd.C0          # after the incident pulse has passed the probe
    refl = before[late][np.argmax(np.abs(before[late]))]
    return refl / inc, after.max() / inc, res

measured = {}
for n2 in (2.0, 1.5):
    r_m, t_m, res = step_run(n2**2)
    measured[n2] = (r_m, t_m)
    r_f, t_f = (1 - n2) / (1 + n2), 2 / (1 + n2)
    print(f"n2 = {n2}: measured r = {r_m:+.4f} (formula {r_f:+.4f}),  t = {t_m:.4f} (formula {t_f:.4f}),  R + T = {r_m**2 + n2 * t_m**2:.4f}")
    if n2 == 2.0:
        res2 = res

fig, ax = plt.subplots(figsize=(11, 3.8))
xg_cm = 100 * res2["x"]
for k_snap, col in ((2, "C0"), (4, "C1"), (6, "C2")):
    ax.plot(xg_cm, res2["snapshots"][k_snap], color=col, label=f"step {(k_snap + 1) * 500}")
ax.axvspan(140, 200, color="grey", alpha=0.15); ax.text(150, 0.85, "eps_r = 4  (n = 2)")
ax.set_xlabel("x [cm]"); ax.set_ylabel("E_y"); ax.legend(loc="upper left")
ax.set_title("a pulse meets a dielectric: reflected -1/3, transmitted 2/3, half as fast and half as wide"); plt.show()

print("\nskin depth delta = sqrt(2 / (omega mu0 sigma))")
for name, sigma, freqs in (("copper", 5.96e7, (50.0, 1e6, 1e9)), ("seawater", 4.0, (1e4, 1e6))):
    for fr in freqs:
        delta = np.sqrt(2 / (2 * np.pi * fr * MU0 * sigma))
        print(f"   {name:9s} f = {fr:8.1e} Hz:  delta = {delta:.3e} m")
```

### 3.7 Radiation

Static charges make static fields, and steady currents make steady fields. Neither radiates. A wave needs a source that changes in time, and the essential ingredient is acceleration: a charge moving at constant velocity carries its field along, but a charge that is accelerated sends out a disturbance in its field that travels away at $c$. An antenna is a conductor in which an oscillating voltage drives charges back and forth, so they accelerate continually. The simplest is a short straight wire fed at its centre, a dipole antenna: its charge distribution is an oscillating electric dipole.

**Why radiation fields fall as $1/r$.** Close to the antenna the fields look like the static fields of module 1, a dipole field falling as $1/r^3$. Far away, more than a wavelength or so, a different part of the field takes over. Its form follows from energy conservation alone. In the far field the waves move outward and the energy crossing every sphere around the source per second is the same, the radiated power $P$. A sphere has area $4\pi r^2$, so the intensity falls as $I = P/4\pi r^2$ for an isotropic source, and since $I \propto E^2$ the field falls as $E \propto 1/r$. That slow fall-off is why radio works over thousands of kilometres while static fields are undetectable a few metres from their source. A real dipole antenna is not isotropic: it radiates most strongly perpendicular to its wire, with intensity proportional to $\sin^2\theta$, and not at all along it.

**The two-dimensional analogue.** Notebook 04 computes in 2D, where nothing varies along $z$. A point source there is really a line source extending along $z$, and the waves are cylinders, not spheres. The power per unit length crossing a cylinder of radius $r$ is constant, the circumference is $2\pi r$, so $I \propto 1/r$ and $E \propto 1/\sqrt{r}$. Doubling the distance lowers the field by $\sqrt2 = 1.414$, the ratio notebook 04 measures between two probes.

**Worked example 13 (a 1 kW transmitter).** An isotropic source of 1 kW gives at 1 km $I = 1000/(4\pi\times10^6) = 7.96\times10^{-5}\,\mathrm{W/m^2}$ and $E_0 = \sqrt{2\times376.73\times7.96\times10^{-5}} = 0.245\,$V/m. At 10 km the field is ten times smaller, 24.5 mV/m, and the intensity a hundred times smaller. A line source radiating the same power per metre of length would lose only a factor $\sqrt{10} = 3.16$ in field over the same tenfold distance.

```python
P, P_line = 1000.0, 1000.0                                    # W, and W per metre of a line source for the 2D case
E0_point = lambda r: np.sqrt(2 * eta0 * P / (4 * np.pi * r**2))     # intensity P / (4 pi r^2) over a sphere
E0_line = lambda r: np.sqrt(2 * eta0 * P_line / (2 * np.pi * r))    # intensity P_line / (2 pi r) over a cylinder
for rr in (1e3, 1e4):
    print(f"isotropic 1 kW at r = {rr:7.0f} m:  I = {P / (4 * np.pi * rr**2):.3e} W/m^2,  E0 = {E0_point(rr):.4f} V/m")
print(f"field ratio from 1 km to 10 km: point source {E0_point(1e3) / E0_point(1e4):.2f},  line source {E0_line(1e3) / E0_line(1e4):.2f}")
print(f"field ratio for a doubled distance, line source: {E0_line(1e3) / E0_line(2e3):.4f} = sqrt(2)")
r = np.logspace(1, 4, 50)                                     # 10 m to 10 km
E0_3d, E0_2d = E0_point(r), E0_line(r)
plt.loglog(r, E0_3d, label="point source in 3D: E ~ 1/r")
plt.loglog(r, E0_2d, label="line source (2D): E ~ 1/sqrt(r)")
plt.xlabel("r [m]"); plt.ylabel("E0 [V/m]"); plt.legend(); plt.title("energy conservation sets the far-field fall-off"); plt.show()
```

### Exercise: Fresnel coefficients at normal incidence

Replace the body of `fresnel_normal` with your own derivation of the amplitude and intensity coefficients for a wave going from medium $n_1$ into medium $n_2$. The check requires $R + T = 1$ for several pairs, including the less-dense case where $t > 1$, and agreement within 1 percent with the coefficients measured by the 1D solver in section 3.6.

```python
def fresnel_normal(n1, n2):
    '''Amplitude coefficients r, t and power fractions R, T for normal incidence from n1 into n2.'''
    r = (n1 - n2) / (n1 + n2)                       # replace from here with your own version
    t = 2 * n1 / (n1 + n2)
    R = r**2
    T = (n2 / n1) * t**2
    return r, t, R, T


for n1, n2 in ((1.0, 1.5), (1.5, 1.0), (1.0, 1.33), (1.0, 2.0), (2.0, 1.0)):
    r_c, t_c, R_c, T_c = fresnel_normal(n1, n2)
    print(f"n1 = {n1}, n2 = {n2}:  r = {r_c:+.4f}  t = {t_c:.4f}  R = {R_c:.4f}  T = {T_c:.4f}  R + T = {R_c + T_c:.12f}")
    assert np.isclose(R_c + T_c, 1.0)
for n2, (r_m, t_m) in measured.items():
    r_c, t_c, _, _ = fresnel_normal(1.0, n2)
    assert abs(r_m / r_c - 1) < 0.01 and abs(t_m / t_c - 1) < 0.01
print("exercise passed")
```

| Gauss | $\nabla\cdot\mathbf{E} = \rho/\epsilon_0$ | always |
| no magnetic charge | $\nabla\cdot\mathbf{B} = 0$ | always |
| Faraday | $\nabla\times\mathbf{E} = -\partial\mathbf{B}/\partial t$ | always |
| Ampere-Maxwell | $\nabla\times\mathbf{B} = \mu_0\mathbf{J} + \mu_0\epsilon_0\,\partial\mathbf{E}/\partial t$ | always |
| Lorentz force | $\mathbf{F} = q(\mathbf{E} + \mathbf{v}\times\mathbf{B})$ | point charge |
| charge conservation | $\nabla\cdot\mathbf{J} + \partial\rho/\partial t = 0$ | follows from the equations |
| wave equation | $\nabla^2\mathbf{E} = c^{-2}\,\partial^2\mathbf{E}/\partial t^2$, same for $\mathbf{B}$ | vacuum, no sources |
| speed of light | $c = 1/\sqrt{\mu_0\epsilon_0} = 2.998\times10^8$ m/s | vacuum |
| d'Alembert solution | $f(x - ct) + g(x + ct)$ | 1D wave equation |
| plane wave | $\mathbf{E} = E_0\cos(kx - \omega t)\hat{\mathbf{y}}$, $\mathbf{B} = (E_0/c)\cos(kx - \omega t)\hat{\mathbf{z}}$, $\omega = ck$ | vacuum |
| transversality | $\mathbf{B} = \hat{\mathbf{k}}\times\mathbf{E}/c$, both perpendicular to $\hat{\mathbf{k}}$ | plane wave in vacuum |
| energy density | $u = \epsilon_0E^2/2 + B^2/2\mu_0$; equal halves in a wave | always; halves for a plane wave |
| Poynting's theorem | $\partial u/\partial t + \nabla\cdot\mathbf{S} = -\mathbf{J}\cdot\mathbf{E}$ | always |
| Poynting vector | $\mathbf{S} = \mathbf{E}\times\mathbf{B}/\mu_0 = \mathbf{E}\times\mathbf{H}$ | vacuum ($\mathbf{H} = \mathbf{B}/\mu_0$) |
| impedance of free space | $\eta_0 = \sqrt{\mu_0/\epsilon_0} = 376.73\,\Omega$, $H = E/\eta_0$ | plane wave in vacuum |
| intensity | $I = E_0^2/2\eta_0 = c\epsilon_0E_0^2/2$; pressure $I/c$ if absorbed | sinusoidal plane wave |
| speed in a dielectric | $v = c/n$, $n = \sqrt{\epsilon_r}$, $\eta = \eta_0/n$ | linear, non-magnetic |
| normal incidence | $r = (n_1 - n_2)/(n_1 + n_2)$, $t = 2n_1/(n_1 + n_2)$, $R = r^2$, $T = (n_2/n_1)t^2$ | sharp boundary, no surface current |
| skin depth | $\delta = \sqrt{2/\omega\mu_0\sigma}$, field $\propto e^{-x/\delta}$ | good conductor, $\sigma \gg \epsilon_0\omega$ |
| far-field fall-off | $E \propto 1/r$ (3D point source), $E \propto 1/\sqrt{r}$ (2D line source) | more than a wavelength away |

## 7. Warm-up answers

1. A flat surface on the loop is pierced by the wire and collects $I$, but a bag-shaped surface with the same rim passing between the plates collects no conduction current. Ampere's law would then give two different values for one circulation. The flaw is that charge accumulates on the plates, so $\nabla\cdot\mathbf{J} \neq 0$.
2. $\nabla\cdot\mathbf{J} + \partial\rho/\partial t = 0$. Where charge piles up, $\partial\rho/\partial t > 0$, so more current flows in than out and $\nabla\cdot\mathbf{J} < 0$.
3. Gauss: $\nabla\cdot\mathbf{E} = \rho/\epsilon_0$. No magnetic charge: $\nabla\cdot\mathbf{B} = 0$. Faraday: $\nabla\times\mathbf{E} = -\partial\mathbf{B}/\partial t$. Ampere-Maxwell: $\nabla\times\mathbf{B} = \mu_0\mathbf{J} + \mu_0\epsilon_0\,\partial\mathbf{E}/\partial t$.
4. $c = 1/\sqrt{\mu_0\epsilon_0} = 2.998\times10^8$ m/s, the speed of light, built from two constants measured with charges and currents at rest.
5. Along $z$, so that $\mathbf{E}\times\mathbf{B}$ points along $+x$. In phase with $\mathbf{E}$, with $B = E/c$.
6. The energy flux of the fields, energy per unit time per unit area crossing a surface perpendicular to it, in $\mathrm{W/m^2}$. $\mathbf{S} = \mathbf{E}\times\mathbf{B}/\mu_0$.
7. $r = (1 - 1.5)/(1 + 1.5) = -0.2$, so $R = r^2 = 4\%$ of the intensity is reflected and 96 percent is transmitted.
