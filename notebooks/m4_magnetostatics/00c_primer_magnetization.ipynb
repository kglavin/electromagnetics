## 2. The ideas

**Atoms are current loops.** Every electron in an atom is a moving charge. Its orbital motion is a tiny current loop, and the electron also carries an intrinsic spin that behaves like a second tiny loop. Primer 00a showed that a current loop of area $A$ carrying current $I$ has a magnetic moment $\mathbf{m} = IA\,\hat{\mathbf{n}}$, feels a torque $\mathbf{m}\times\mathbf{B}$ that turns it into alignment with the field, and produces, far away, the field of a magnetic dipole. An atom is therefore a magnetic dipole, with a moment of the order of the Bohr magneton $\mu_B = 9.27\times10^{-24}\,$J/T. In most atoms the electron moments pair off and cancel, leaving no net moment; in some (iron, cobalt, nickel, many rare-earth ions) a net moment of a few $\mu_B$ survives.

**Magnetization.** When the atomic moments in a piece of material point in a common direction, either because a field has aligned them or because the material does it spontaneously, the material has a net magnetic moment. The magnetization $\mathbf{M}$ is that moment per unit volume, averaged over a region large compared with an atom and small compared with the body. Its units are A m$^2$/m$^3$ = A/m, the same as current per unit length. This is the magnetic twin of the polarization $\mathbf{P}$ of module 3: there, dipole moment per volume with units of surface charge; here, magnetic moment per volume with units of surface current. The parallel is exact, and every result below has an electric counterpart in primer M3.0a.

**A magnetized body is a set of currents.** Picture a slab filled with identical small current loops, all circulating the same way. Inside the slab, where two neighbouring loops touch, their currents run in opposite directions and cancel. At the outer edge of the slab there is no neighbour to cancel against, so a net current flows around the rim. A uniformly magnetized body is therefore equivalent to a surface current $\mathbf{K}_b = \mathbf{M}\times\hat{\mathbf{n}}$ on its boundary and nothing inside. If $\mathbf{M}$ varies from place to place, neighbouring loops no longer cancel exactly and a net current $\mathbf{J}_b = \nabla\times\mathbf{M}$ is left in the volume as well. These are the bound currents. They are called bound because the electrons that carry them never leave their atoms, but they are real currents: they produce a magnetic field through the Biot-Savart law exactly as a wire does, and that field is the macroscopic $\mathbf{B}$ of the magnetized body. The bar magnet on a refrigerator is, for every purpose in this course, a solenoid whose current happens to be carried by electrons circulating inside iron atoms.

**The auxiliary field $\mathbf{H}$.** Ampere's law, $\nabla\times\mathbf{B} = \mu_0\mathbf{J}$, counts all currents: the free currents $\mathbf{J}_f$ we drive through wires and the bound currents the material produces in response. The bound currents depend on $\mathbf{M}$, which depends on $\mathbf{B}$, which is what we want to find. Moving the bound part to the left-hand side defines a new field, $\mathbf{H} = \mathbf{B}/\mu_0 - \mathbf{M}$, with $\nabla\times\mathbf{H} = \mathbf{J}_f$: its circulation counts free current only. This is the twin of $\mathbf{D}$ in module 3. Like $\mathbf{D}$ it is a bookkeeping field, with no force law of its own; unlike $\mathbf{B}$ its divergence is not zero, and the places where $\mathbf{M}$ starts and stops act as sources of $\mathbf{H}$. Those sources are the magnetic poles of the old picture of magnets. Both pictures, currents for $\mathbf{B}$ and poles for $\mathbf{H}$, are exact; they describe the same field from two sides.

**Three kinds of material.** How much magnetization a field produces depends on the atoms. In a diamagnet every atom has zero net moment, and the applied field induces a tiny moment pointing against it: $\mathbf{M}$ is opposite to $\mathbf{B}$, and the effect is minute, with a susceptibility of order $-10^{-5}$ (water, copper, bismuth). In a paramagnet the atoms carry permanent moments that thermal motion scatters at random; a field aligns a small fraction of them, giving a weak $\mathbf{M}$ along the field, susceptibility of order $+10^{-5}$ to $10^{-3}$ (aluminium, oxygen, most salts of transition metals). In a ferromagnet (iron, cobalt, nickel) the moments align with each other spontaneously over regions called domains; an applied field grows the favourably aligned domains at the expense of the others, and $\mathbf{M}$ can become enormous, up to $1.7\times10^6\,$A/m for iron, equivalent to more than two tesla. Ferromagnets are the only magnetic materials whose effects are large, and they are also the ones for which the response is non-linear and depends on history (hysteresis).

**Boundary conditions.** Where the material changes, the field changes abruptly, and two rules say how. Because $\nabla\cdot\mathbf{B} = 0$ always, the normal component of $\mathbf{B}$ is continuous at any interface. Because $\nabla\times\mathbf{H} = \mathbf{J}_f$, the tangential component of $\mathbf{H}$ is continuous unless a free surface current flows along the interface. Across the surface of a magnet, where $\mathbf{K}_b$ flows, the tangential component of $\mathbf{B}$ therefore jumps by $\mu_0 K_b$. These are the twins of the dielectric rules (normal $\mathbf{D}$ continuous, tangential $\mathbf{E}$ continuous), and they make field lines refract at the surface of iron so strongly that iron appears to capture them.

**Magnetic circuits.** The high permeability of iron lets a ring of iron guide magnetic flux the way a copper wire guides current, and there is a one-to-one analogy: flux plays the role of current, the ampere-turns $NI$ of a winding play the role of emf, and each piece of the path has a reluctance $l/\mu A$ that plays the role of resistance. Transformers, motors and electromagnets are designed with it, and it shows why a small air gap dominates everything: air has a permeability a thousand times smaller than iron, so a millimetre of air has the reluctance of a metre of iron.

Everything it relies on is derived below.

### 3.1 Magnetic dipoles in matter

Primer 00a established the magnetic moment of a planar current loop, $\mathbf{m} = IA\,\hat{\mathbf{n}}$ with $\hat{\mathbf{n}}$ the normal given by the right-hand rule, the torque $\boldsymbol{\tau} = \mathbf{m}\times\mathbf{B}$ it feels in a uniform field, its energy $U = -\mathbf{m}\cdot\mathbf{B}$, and the dipole field it produces far away, $\mathbf{B} = \dfrac{\mu_0}{4\pi}\dfrac{3(\mathbf{m}\cdot\hat{\mathbf{r}})\hat{\mathbf{r}} - \mathbf{m}}{r^3}$. All of that applies to atoms.

An electron of charge $-e$ orbiting a nucleus is a current loop, and the moment of that loop turns out to be proportional to the orbital angular momentum, $\mathbf{m} = -\dfrac{e}{2m_e}\mathbf{L}$. Quantum mechanics restricts $L$ to multiples of $\hbar$, so the natural unit of atomic moment is the Bohr magneton,
$$\mu_B = \frac{e\hbar}{2m_e} = 9.27\times10^{-24}\,\mathrm{J/T} \ (=\mathrm{A\,m^2}).$$
The electron's spin carries a moment of almost exactly one $\mu_B$ as well. In a filled electron shell the moments cancel pairwise. The atoms that matter magnetically are those with unpaired electrons: iron has four, giving a free-atom moment of about $4\mu_B$, which the metallic environment reduces to about $2.2\mu_B$ per atom.

The magnetization is the net moment per unit volume,
$$\mathbf{M} = \frac{\sum\mathbf{m}_i}{\Delta v},$$
with the sum over the atoms in a volume $\Delta v$ that is tiny on our scale and huge on the atomic scale. Units: A m$^2$ / m$^3$ = A/m. Keep the units in mind; they are the first hint that a magnetization is secretly a current per unit length.

**Worked example 1 (iron).** Iron has density $7874\,$kg/m$^3$ and molar mass $55.85\,$g, so its number density is $n = 7874\times6.022\times10^{23}/0.05585 = 8.49\times10^{28}\,$atoms/m$^3$. Fully magnetized iron (every atomic moment aligned) has $M_{\text{sat}} = 1.71\times10^6\,$A/m. The moment per atom is then $M_{\text{sat}}/n = 2.01\times10^{-23}\,$J/T $= 2.17\,\mu_B$. That is the number quoted above, obtained from a macroscopic measurement.

**Worked example 2 (a small magnet).** A $1\,$cm$^3$ block with $M = 8\times10^5\,$A/m (a typical ferrite or an iron magnet well below saturation) has moment $m = MV = 8\times10^5\times10^{-6} = 0.8\,$A m$^2$. If that moment were a single loop of area $1\,$cm$^2$, the loop would carry $I = m/A = 0.8/10^{-4} = 8000\,$A. In a field of $0.1\,$T the maximum torque is $mB = 0.08\,$N m. On its axis at $10\,$cm, far enough to use the dipole formula, the field is $B = \dfrac{\mu_0}{4\pi}\dfrac{2m}{r^3} = 10^{-7}\times\dfrac{1.6}{10^{-3}} = 1.6\times10^{-4}\,$T, about three times the Earth's field.

```python
e, hbar, m_e = 1.602e-19, 1.0546e-34, 9.109e-31
mu_B = e * hbar / (2 * m_e)
n_iron = 7874 * 6.022e23 / 0.05585
M_sat = 1.71e6
print(f"Bohr magneton: {mu_B:.3e} J/T")
print(f"iron: {n_iron:.3e} atoms/m^3, moment per atom {M_sat / n_iron:.3e} J/T = {M_sat / n_iron / mu_B:.2f} mu_B")

M, V = 8e5, 1e-6
m = M * V
print(f"\n1 cm^3 block with M = {M:.0e} A/m: m = {m:.2f} A m^2, equivalent single-loop current {m / 1e-4:.0f} A, max torque in 0.1 T: {m * 0.1:.3f} N m")
B_axis = bs.dipole_B([0, 0, m], np.array([[0.0, 0.0, 0.1]]))[0, 2]
print(f"dipole field 10 cm along the axis: {B_axis:.2e} T   (Earth's field is about 5e-5 T)")
```

### 3.2 Bound currents

Why is a magnetized body equivalent to currents? Take a thin slab, uniformly magnetized along $z$, and picture it as a tiling of identical small square loops in the $xy$ plane, each of side $a$, thickness $t$, carrying current $I$ around its edge so that its moment $Ia^2$ points along $z$. The magnetization is moment over volume, $M = Ia^2/(a^2t) = I/t$. Now look at any internal edge: it belongs to two adjacent loops, and the two currents on that edge run in opposite directions and cancel. Only the edges on the outer boundary of the slab carry an uncancelled current $I$. That current, spread over the slab thickness $t$, is a surface current of density $K_b = I/t = M$, flowing around the rim in the direction $\hat{\mathbf{M}}\times\hat{\mathbf{n}}$. In general,
$$\mathbf{K}_b = \mathbf{M}\times\hat{\mathbf{n}},$$
with $\hat{\mathbf{n}}$ the outward normal of the surface. On a face whose normal is parallel to $\mathbf{M}$ (the top and bottom of the slab) the cross product vanishes: no current there.

If $\mathbf{M}$ is not uniform the cancellation inside is incomplete. Consider two adjacent loops at $x$ and $x + a$ whose moments along $z$ differ: $M_z(x + a) > M_z(x)$. On their shared edge, parallel to $y$, the currents are $I(x + a)$ one way and $I(x)$ the other, leaving a net current $[M_z(x + a) - M_z(x)]\,t$ flowing along $-y$. Per unit area of the $xz$ cross-section that is $-\partial M_z/\partial x$ in the $y$ direction. Collecting the same argument for every pair of components gives
$$\mathbf{J}_b = \nabla\times\mathbf{M}.$$
The same two formulas follow from a longer route, which the book takes: write the vector potential of a continuous distribution of dipoles and integrate by parts, and the volume term is $\nabla\times\mathbf{M}$ while the surface term is $\mathbf{M}\times\hat{\mathbf{n}}$. Compare module 3: $\rho_b = -\nabla\cdot\mathbf{P}$ and $\sigma_b = \mathbf{P}\cdot\hat{\mathbf{n}}$, derived by the same counting argument with charges in place of currents.

Two consequences. First, $\nabla\cdot\mathbf{J}_b = \nabla\cdot(\nabla\times\mathbf{M}) = 0$: bound current never piles up charge anywhere, which it must not, since the electrons stay with their atoms. Second, the total bound current flowing through any cross-section of a body is zero: what goes one way in the volume comes back along the surface. The bound currents form closed loops, just as the total bound charge of a dielectric is zero.

**Worked example 3 (uniformly magnetized cylinder).** A cylinder of radius $R$ and length $L$ with $\mathbf{M} = M\hat{\mathbf{z}}$ along its axis. Volume current: $\nabla\times\mathbf{M} = 0$. On the flat ends $\hat{\mathbf{n}} = \pm\hat{\mathbf{z}}$ and $\mathbf{M}\times\hat{\mathbf{n}} = 0$. On the curved surface $\hat{\mathbf{n}} = \hat{\mathbf{r}}$ (cylindrical) and $\mathbf{K}_b = M\hat{\mathbf{z}}\times\hat{\mathbf{r}} = M\hat{\boldsymbol{\phi}}$: a sheet current of density $M$ circulating around the cylinder. That is exactly a solenoid with $nI = M$ (primer 00b). For a cylinder much longer than its radius the field inside is therefore $B = \mu_0 nI = \mu_0M$ and the field outside is nearly zero. With $M = 8\times10^5\,$A/m: $B = 1.005\,$T. A long bar magnet is a solenoid.

**Worked example 4 (uniformly magnetized sphere).** $\mathbf{M} = M\hat{\mathbf{z}}$ inside radius $R$. No volume current. On the surface $\hat{\mathbf{n}} = \hat{\mathbf{r}}$ (spherical), and $\hat{\mathbf{z}}\times\hat{\mathbf{r}} = \sin\theta\,\hat{\boldsymbol{\phi}}$, so $\mathbf{K}_b = M\sin\theta\,\hat{\boldsymbol{\phi}}$: strongest at the equator, zero at the poles, circulating about the $z$ axis. This is the current distribution of a spinning uniformly charged shell, and its field inside is uniform, $\mathbf{B} = \tfrac23\mu_0\mathbf{M}$ (the exercise builds it from loops and checks the $\tfrac23$). Outside it is the pure dipole field of $\mathbf{m} = \tfrac43\pi R^3\mathbf{M}$. Compare the polarized sphere of module 3, $\mathbf{E} = -\mathbf{P}/3\epsilon_0$: the magnetic field inside a magnetized sphere points along $\mathbf{M}$, the electric field inside a polarized sphere points against $\mathbf{P}$. The sign difference is the difference between the field of a current loop (along its moment inside the loop) and the field of a charge pair (against its moment between the charges).

**Worked example 5 (a non-uniform magnetization).** A cylinder of radius $R$ with $\mathbf{M} = k\,r\,\hat{\mathbf{z}}$, growing linearly from the axis. Volume current, in cylindrical coordinates: $(\nabla\times\mathbf{M})_\phi = -\partial M_z/\partial r = -k$, so $\mathbf{J}_b = -k\hat{\boldsymbol{\phi}}$, a uniform current density circulating the wrong way round. Surface current on the curved face: $\mathbf{K}_b = M(R)\hat{\mathbf{z}}\times\hat{\mathbf{r}} = kR\,\hat{\boldsymbol{\phi}}$. Total current through a half-plane cut from the axis to the surface, per unit length: volume $\int_0^R(-k)\,dr = -kR$, surface $+kR$, total zero. With $k = 10^6\,$A/m$^2$ and $R = 0.8\,$m: $J_b = -10^6\,$A/m$^2$ and $K_b = 8\times10^5\,$A/m.

The code puts two magnetized cylinders on a 2D grid. First a uniform one, $\mathbf{M} = M_0\hat{\mathbf{z}}$ inside $r = R$: the curl is zero inside, and at the surface the step of $M_z$ to zero produces a spike one or two cells wide whose integral across the surface is $K_b = M_0$, the same mechanism that produced $\sigma_b$ from a step in $\mathbf{P}$ in module 3. Then worked example 5: the curl gives $-k$ everywhere inside except in the cells next to the axis, where $M_z = kr$ has a kink that a central difference cannot resolve, and the total current through a radial cut, surface spike plus volume, integrates to zero to within a percent of $K_b$, the residue coming from the kink at the axis. For the sphere it evaluates $\mathbf{K}_b = \mathbf{M}\times\hat{\mathbf{n}}$ at a few points.

**Animation.** The counting argument of this section, step by step. Atomic current loops with random senses have no net moment. Aligned, every loop runs counter-clockwise and $\mathbf{M}$ points out of the page. On every shared edge the two currents are equal and opposite and cancel, leaving one current around the rim: $\mathbf{K}_b = \mathbf{M}\times\hat{\mathbf{n}}$, a solenoid with $nI = M$. In the second part $M_z$ grows from left to right: horizontal shared edges still cancel, but each vertical one keeps a net downward current, which is $\mathbf{J}_b = \nabla\times\mathbf{M} = -(\partial M_z/\partial x)\hat{\mathbf{y}}$.

```python
from emlib import show
show.animation("m4_magnetostatics", "BoundCurrents")
```

```python
k, R = 1e6, 0.8
x, y, P = grids.grid2d((-1.2, 1.2), (-1.2, 1.2), 241)
hx, hy = grids.spacing(x, y)
r = np.hypot(P[..., 0], P[..., 1])
phi_hat = np.stack([-P[..., 1] / np.where(r > 0, r, 1), P[..., 0] / np.where(r > 0, r, 1), 0 * r], axis=-1)

j = P.shape[1] // 2                                               # the row y = 0, a radial cut along +x
band = (x > 0.95 * R) & (x < 1.05 * R)

# Uniform magnetization: no volume current, a surface spike that integrates to K_b = M0.
M0 = 8e5
M_uni = np.zeros(P.shape); M_uni[..., 2] = np.where(r <= R, M0, 0.0)
J_uni = np.einsum("ijk,ijk->ij", mg.bound_volume_current(M_uni, hx, hy), phi_hat)
print(f"uniform M: max |J_b| inside (r < 0.9R) = {np.abs(J_uni[r < 0.9 * R]).max():.1e} A/m^2")
print(f"           integral of the surface spike across r = R: {np.trapezoid(J_uni[band, j], dx=hx):.4e} A/m   formula K_b = M0 = {M0:.1e} A/m")

# Worked example 5: M = k r z_hat.
M = np.zeros(P.shape)
M[..., 2] = np.where(r <= R, k * r, 0.0)
J_b = mg.bound_volume_current(M, hx, hy)
J_phi = np.einsum("ijk,ijk->ij", J_b, phi_hat)                  # component along phi_hat
inside = (r > 0.2 * R) & (r < 0.9 * R)                           # away from the kink on the axis and from the surface
print(f"\nM = k r: J_b . phi_hat inside: mean {J_phi[inside].mean():.4e}, spread {J_phi[inside].std():.1e}   formula -k = {-k:.1e} A/m^2")
cut = x >= 0
print(f"           total current through the radial cut (volume + surface): {np.trapezoid(J_phi[cut, j], dx=hx):.2e} A/m   compared with K_b = k R = {k * R:.1e} A/m: zero")

# Sphere: K_b = M x n at several polar angles.
theta = np.radians([0, 30, 60, 90, 120, 180])
normal = np.stack([np.sin(theta), 0 * theta, np.cos(theta)], axis=-1)        # points in the xz plane, phi = 0
K_sphere = mg.bound_surface_current(np.array([0.0, 0.0, 8e5]), normal)
for th, Kv in zip(np.degrees(theta), K_sphere):
    print(f"theta = {th:5.0f} deg: K_b = {Kv}  |K_b| = {np.linalg.norm(Kv):.3e}   M sin(theta) = {8e5 * np.sin(np.radians(th)):.3e}")

fig, ax = plt.subplots(figsize=(6, 5))
im = ax.pcolormesh(x, y, J_phi.T, cmap="RdBu_r", vmin=-3 * k, vmax=3 * k, shading="auto")
ax.add_patch(plt.Circle((0, 0), R, fill=False, color="k", lw=1)); ax.set_aspect("equal")
plt.colorbar(im, label="J_b . phi_hat  [A/m^2]"); ax.set_title("M = k r z_hat: uniform -k inside, a surface spike at r = R"); plt.show()
```

### 3.3 The field of a magnetized body

Once the bound currents are known, the field they produce is pure primer 00a: Biot-Savart superposition, with the surface current $\mathbf{K}_b$ treated as a sheet of thin wires. There is no new law. And the field so obtained is the macroscopic $\mathbf{B}$, the average of the microscopic field over a volume containing many atoms, for the same reason as in module 3: the bound currents are themselves averages over many atomic loops. Summing the dipole fields of individual atoms instead would give the microscopic field, which swings wildly from one lattice site to the next and is not what a Hall probe measures.

The code builds the surface current of a finite uniformly magnetized cylinder as a stack of $N$ circular loops along its length, each carrying the current $K_b\,\Delta z = M\,\Delta z$ of the slice of sheet it replaces. The field on the axis must then match the solenoid formula from primer 00b with $nI = M$:
$$B_z(z) = \frac{\mu_0M}{2}\left[\frac{L/2 + z}{\sqrt{R^2 + (L/2 + z)^2}} + \frac{L/2 - z}{\sqrt{R^2 + (L/2 - z)^2}}\right].$$

**Worked example 6 (two bar magnets).** $M = 8\times10^5\,$A/m, radius $R = 0.1\,$m. A long magnet, $L = 1\,$m: at the centre each bracket term is $0.5/\sqrt{0.01 + 0.25} = 0.9806$, so $B = 0.9806\,\mu_0M = 0.986\,$T, within two percent of the infinite-solenoid value $\mu_0M = 1.005\,$T. A short magnet, $L = 0.2\,$m: each term is $0.1/\sqrt{0.01 + 0.01} = 0.7071$, so $B = 0.711\,$T. At the end face of either magnet ($z = L/2$) the second term vanishes and the first is close to one for the long magnet: $B \approx \mu_0M/2$, half the interior value, the familiar result that the field at the mouth of a long solenoid is half the field deep inside.

```python
def magnet_loops(M, R, L, n_loops=200, n_seg=120):
    '''Surface current of a cylinder magnetized along z, as a stack of loops; returns (mid, dl, I_per_loop).'''
    dz = L / n_loops
    zc = -L / 2 + (np.arange(n_loops) + 0.5) * dz
    mids, dls = zip(*[bs.segments(bs.loop(R, n_seg, centre=(0.0, 0.0, z))) for z in zc])
    return np.vstack(mids), np.vstack(dls), M * dz

M, R = 8e5, 0.1
z = np.linspace(-1.0, 1.0, 401)
axis = np.stack([0 * z, 0 * z, z], axis=-1)
fig, ax = plt.subplots(figsize=(7, 4.5))
for L, colour in ((1.0, "C0"), (0.2, "C3")):
    mid, dl, I = magnet_loops(M, R, L)
    Bz = bs.wire_B(mid, dl, I, axis)[:, 2]
    exact = bs.solenoid_axis_B(R, L, 1.0, M, z)                   # n I = M
    ax.plot(z, Bz, colour, lw=2, label=f"loops, L = {L}")
    ax.plot(z, exact, "k--", lw=1)
    print(f"L = {L}: B at the centre {Bz[200]:.4f} T, solenoid formula {exact[200]:.4f} T, mu0 M = {MU0 * M:.4f} T, at the end face {bs.wire_B(mid, dl, I, np.array([[0, 0, L / 2]]))[0, 2]:.4f} T")
ax.axhline(MU0 * M, color="grey", ls=":", label="mu0 M"); ax.set_xlabel("z [m]"); ax.set_ylabel("B_z on the axis [T]")
ax.legend(); ax.set_title("magnetized cylinders as loop stacks (dashed: solenoid formula with nI = M)"); plt.show()
```

```python
# Field map of the short magnet in the xz plane, with field lines.
L = 0.2
mid, dl, I = magnet_loops(M, R, L, n_loops=60, n_seg=120)
xs = np.linspace(-0.4, 0.4, 81); zs = np.linspace(-0.4, 0.4, 81)
Xg, Zg = np.meshgrid(xs, zs, indexing="ij")
pts = np.stack([Xg, 0 * Xg, Zg], axis=-1)
B = bs.wire_B(mid, dl, I, pts)
Bmag = np.linalg.norm(B, axis=-1)

fig, ax = plt.subplots(figsize=(6.5, 6.5))
ax.streamplot(xs, zs, B[..., 0].T, B[..., 2].T, color=np.log10(Bmag).T, cmap="viridis", density=1.6)
ax.add_patch(plt.Rectangle((-R, -L / 2), 2 * R, L, fill=False, color="C3", lw=2))
ax.set_aspect("equal"); ax.set_xlabel("x"); ax.set_ylabel("z"); ax.set_title("bar magnet: field of its bound surface current (colour = log10 |B|)"); plt.show()
print("inside the magnet the field runs along M (bottom to top); outside it returns from the north face to the south face")
```

### 3.4 $\mathbf{H}$ and Ampere's law inside matter

The total current density is $\mathbf{J} = \mathbf{J}_f + \mathbf{J}_b$: free current, which we drive through wires and control, plus bound current, which the material produces. Ampere's law in differential form (primer 00b) counts both,
$$\nabla\times\mathbf{B} = \mu_0(\mathbf{J}_f + \mathbf{J}_b) = \mu_0\mathbf{J}_f + \mu_0\nabla\times\mathbf{M}.$$
Move the bound term across and divide by $\mu_0$:
$$\nabla\times\left(\frac{\mathbf{B}}{\mu_0} - \mathbf{M}\right) = \mathbf{J}_f.$$
The combination in brackets is given its own name,
$$\mathbf{H} = \frac{\mathbf{B}}{\mu_0} - \mathbf{M}, \qquad \nabla\times\mathbf{H} = \mathbf{J}_f, \qquad \oint\mathbf{H}\cdot d\mathbf{l} = I_{f,\text{enc}}.$$
$\mathbf{H}$ has the units of $\mathbf{M}$, A/m, and Ampere's law for $\mathbf{H}$ involves only the current we know. In a symmetric geometry it gives $\mathbf{H}$ in one line, whatever the material, exactly as the generalized Gauss' law gave $\mathbf{D}$ in module 3.

What $\mathbf{H}$ is not. It is not the field that exerts forces: the Lorentz force uses $\mathbf{B}$. Its divergence is not zero: $\nabla\cdot\mathbf{H} = -\nabla\cdot\mathbf{M}$, because $\nabla\cdot\mathbf{B} = 0$. Wherever $\mathbf{M}$ ends, at the north face of a magnet, $\mathbf{H}$ has a positive source; where $\mathbf{M}$ begins, at the south face, a negative one. These sources are the magnetic poles of the pre-Ampere description, and they are a perfectly good way to compute $\mathbf{H}$: a uniformly magnetized bar has a surface "pole density" $\mathbf{M}\cdot\hat{\mathbf{n}}$ on its end faces, and $\mathbf{H}$ is the field of those two sheets computed as if they were charge sheets with $\epsilon_0 \to 1$. Both pictures describe the same $\mathbf{B}$. Inside a magnet they look very different: the current picture gives $\mathbf{B}$ pointing along $\mathbf{M}$, the pole picture gives $\mathbf{H}$ pointing against $\mathbf{M}$, from north pole to south pole inside the bar. That interior $\mathbf{H}$ is called the demagnetizing field, the twin of the depolarizing field in a dielectric.

**Worked example 7 (a filled solenoid).** A long solenoid with $n = 1000$ turns per metre carries $I = 2\,$A. Ampere's law for $\mathbf{H}$ around the usual rectangle, one side inside and one far outside, gives $H = nI = 2000\,$A/m inside, whatever fills it. Empty: $M = 0$ and $B = \mu_0H = 2.51\,$mT. Filled with a material of magnetization $M$: $B = \mu_0(H + M)$. The winding fixes $H$; the material adds $\mu_0M$ on top.

**Worked example 8 (the demagnetizing field).** The two magnets of worked example 6 have no free current at all, so $\oint\mathbf{H}\cdot d\mathbf{l} = 0$ around every loop and $\mathbf{H}$ is curl-free everywhere. Inside, on the axis, $H_z = B_z/\mu_0 - M$. Long magnet: $B_z = 0.9806\,\mu_0M$ at the centre, so $H_z = -0.0194\,M = -1.55\times10^4\,$A/m. Short magnet: $B_z = 0.7071\,\mu_0M$, so $H_z = -0.293\,M = -2.34\times10^5\,$A/m. The shorter the magnet, the larger the demagnetizing field, because the poles on its end faces are closer. For the limit of a thin disk magnetized across its thickness, $B_z \to 0$ inside and $H_z \to -M$; for a sphere, $B = \tfrac23\mu_0M$ gives $H = -\tfrac13M$, the exact twin of $E = -P/3\epsilon_0$. The fraction $-H/M$ is called the demagnetizing factor: $0$ for a long needle, $1/3$ for a sphere, $1$ for a thin plate.

The code computes $H_z$ along the axis of both magnets from the loop-stack $B_z$. Outside the magnet $M = 0$ and $H = B/\mu_0$ points along the field; inside it reverses.

```python
fig, ax = plt.subplots(figsize=(7, 4.5))
for L, colour in ((1.0, "C0"), (0.2, "C3")):
    mid, dl, I = magnet_loops(M, R, L)
    Bz = bs.wire_B(mid, dl, I, axis)[:, 2]
    Mz = np.where(np.abs(z) < L / 2, M, 0.0)
    Hz = mg.H_field(np.stack([0 * Bz, 0 * Bz, Bz], axis=-1), np.stack([0 * Mz, 0 * Mz, Mz], axis=-1))[:, 2]
    ax.plot(z, Hz / M, colour, lw=2, label=f"H_z / M, L = {L}")
    centre = np.argmin(np.abs(z))
    print(f"L = {L}: H_z at the centre = {Hz[centre]:.3e} A/m = {Hz[centre] / M:+.4f} M   (hand: {(L / 2) / np.hypot(R, L / 2) - 1:+.4f} M)")
ax.axhline(0, color="grey", lw=0.5); ax.set_xlabel("z [m]"); ax.set_ylabel("H_z / M"); ax.legend()
ax.set_title("H inside a magnet points against M: the demagnetizing field"); plt.show()
```

### 3.5 Linear materials and ferromagnets

For diamagnets and paramagnets the magnetization is proportional to the field and small. The convention is to write it in terms of $\mathbf{H}$, because $\mathbf{H}$ is what the free currents fix:
$$\mathbf{M} = \chi_m\mathbf{H}, \qquad \mathbf{B} = \mu_0(\mathbf{H} + \mathbf{M}) = \mu_0(1 + \chi_m)\mathbf{H} = \mu\mathbf{H}, \qquad \mu_r = 1 + \chi_m.$$
$\chi_m$ is the magnetic susceptibility, $\mu$ the permeability, $\mu_r$ the relative permeability. Typical values: diamagnets $\chi_m \approx -10^{-5}$ (water $-9\times10^{-6}$, copper $-1\times10^{-5}$, bismuth $-1.7\times10^{-4}$); paramagnets $\chi_m \approx +10^{-5}$ to $10^{-3}$ (aluminium $+2\times10^{-5}$, liquid oxygen $+4\times10^{-3}$). For all of these $\mu_r$ differs from one by a part in $10^5$ or so, and for practical purposes they are magnetically transparent. The small negative response of diamagnets comes from the orbital currents induced in every atom by a changing field (Faraday's law, module 6, at the atomic scale), which by Lenz's law oppose the field; it is present in all materials and masked in paramagnets by the alignment of permanent moments, which thermal motion limits to a fraction of order $\mu_BB/k_BT$ of full alignment, about $10^{-3}$ at one tesla and room temperature.

Ferromagnets are different in kind. Below a critical temperature (the Curie temperature, $1043\,$K for iron) neighbouring atomic moments lock together by a quantum effect and align spontaneously over regions of micrometre size called domains. In an unmagnetized piece the domains point every which way and cancel. An applied field of a few hundred A/m grows the domains aligned with it at the expense of the others, so $M$ rises to values of order $10^6\,$A/m, a thousand times larger than $H$: an effective $\mu_r$ in the hundreds or thousands. Three things follow that no linear law captures. The response saturates: once every domain is aligned, $M = M_{\text{sat}}$ and no further field helps. The response has memory: when $H$ is removed, part of $M$ remains (remanence), and a reverse field of some size (coercivity) is needed to bring $M$ back to zero, so $M(H)$ traces a loop (hysteresis) rather than a line. And the response depends on the material's history and microstructure. Soft magnetic materials (transformer iron, permalloy) have narrow loops: large $\mu_r$, tiny remanence, easily magnetized and demagnetized, used as cores. Hard materials (alnico, ferrites, neodymium alloys) have wide loops: large remanence and coercivity, so they stay magnetized against their own demagnetizing field, which is what a permanent magnet is.

**Worked example 9 (a solenoid with a core).** The solenoid of worked example 7, $H = 2000\,$A/m. With an aluminium core: $B = \mu_0(1 + 2\times10^{-5})H = 2.5133\,$mT, indistinguishable from empty. With a soft-iron core quoted as $\mu_r = 1000$: nominally $B = \mu_0\mu_rH = 2.51\,$T. But that exceeds the saturation of iron, $\mu_0M_{\text{sat}} = 2.15\,$T, so the linear law has been pushed past its range. The real answer is $B = \mu_0(H + M_{\text{sat}}) = 0.0025 + 2.149 = 2.15\,$T, set by the material, not the winding. Lesson: a quoted $\mu_r$ for iron is a small-field value, and the first thing to check is whether $\mu_rH$ exceeds $M_{\text{sat}}$.

```python
n_turns, I_sol = 1000, 2.0
H = n_turns * I_sol
for name, chi in (("vacuum", 0.0), ("water", -9e-6), ("aluminium", 2e-5), ("liquid oxygen", 4e-3)):
    print(f"{name:14s} chi_m = {chi:+.0e}: M = {chi * H:+.3e} A/m, B = {mg.B_from_H(np.array([H]), chi)[0] * 1e3:.5f} mT")
B_linear = MU0 * 1000 * H
print(f"\niron, mu_r = 1000 taken literally: B = {B_linear:.3f} T, but mu0 M_sat = {MU0 * M_sat:.3f} T, so M would need to be {1000 * H:.1e} A/m > M_sat = {M_sat:.2e}")
print(f"saturated core: B = mu0 (H + M_sat) = {MU0 * (H + M_sat):.3f} T")
```

### 3.6 Boundary conditions

At an interface between two materials the fields on the two sides are related by two rules, each from one of the two magnetostatic equations applied to a vanishingly small region straddling the interface, exactly as in primer M3.0a section 3.6.

Normal $\mathbf{B}$. Take a flat pillbox with its faces parallel to the interface, one in each material, and let its height shrink to zero. $\nabla\cdot\mathbf{B} = 0$ means zero net flux out of it, and the curved side contributes nothing as it shrinks, so the flux in through one face equals the flux out through the other:
$$B_{1n} = B_{2n}.$$
Always, for every interface, because there are no magnetic charges.

Tangential $\mathbf{H}$. Take a thin rectangular loop with its long sides parallel to the interface, one in each material, and let its short sides shrink. Ampere's law for $\mathbf{H}$ says the circulation equals the free current threading the loop, which for a vanishing rectangle is the free surface current $K_f$ flowing along the interface perpendicular to the loop:
$$H_{1t} - H_{2t} = K_f \quad(\text{as a vector: } \hat{\mathbf{n}}\times(\mathbf{H}_1 - \mathbf{H}_2) = \mathbf{K}_f).$$
With no free surface current, tangential $\mathbf{H}$ is continuous. The same loop argument applied to $\mathbf{B}$ with Ampere's law for all currents gives $B_{1t} - B_{2t} = \mu_0(K_f + K_b)$: across the surface of a magnet, where $\mathbf{K}_b = \mathbf{M}\times\hat{\mathbf{n}}$ flows, the tangential component of $\mathbf{B}$ jumps by $\mu_0K_b$ even though nothing free is there. This is what worked example 3 already said: just inside a long magnetized cylinder $B = \mu_0M$, just outside $B \approx 0$, a jump of $\mu_0M = \mu_0K_b$ in the component tangential to the curved surface.

**Worked example 10 (refraction).** Two linear materials, no free current. Let the field in material 1 make angle $\theta_1$ with the normal. Continuity of $B_n$: $B_1\cos\theta_1 = B_2\cos\theta_2$. Continuity of $H_t$: $B_1\sin\theta_1/\mu_1 = B_2\sin\theta_2/\mu_2$. Divide: $\tan\theta_1/\tan\theta_2 = \mu_1/\mu_2$. Field lines crossing from iron ($\mu_r = 1000$) into air: a line at $45^\circ$ to the normal inside the iron leaves at $\tan\theta_{\text{air}} = \tan45^\circ/1000$, so $\theta_{\text{air}} = 0.057^\circ$. Lines leave iron almost exactly perpendicular to its surface, whatever they were doing inside. Going the other way, a line at $10^\circ$ in air bends to $\tan\theta_{\text{iron}} = 1000\tan10^\circ$, $\theta_{\text{iron}} = 89.7^\circ$: inside the iron it runs almost parallel to the surface. That is why iron guides flux: lines that enter it are turned to run along it.

The code checks both rules on the loop-stack magnet. The curved surface has normal $\hat{\mathbf{r}}$, so $B_z$ is tangential there and must jump by $\mu_0M$ while $B_r$ is normal and must be continuous. A stack of polygonal loops reproduces a sheet current only at distances larger than the loop spacing and the segment length, so the samples sit a few of those inside and outside the surface, and the agreement is at the percent level rather than exact.

```python
L, n_loops = 1.0, 500
mid, dl, I = magnet_loops(M, R, L, n_loops=n_loops, n_seg=360)      # loop spacing 2 mm, segments 1.7 mm
dz = L / n_loops
delta = 4 * dz                                                     # a few loop spacings and segment lengths away from the sheet
for z0 in (0.0, 0.25):
    pts = np.array([[R - delta, 0, z0], [R + delta, 0, z0]])
    B_in, B_out = bs.wire_B(mid, dl, I, pts)
    print(f"z = {z0}: tangential B_z inside {B_in[2]:.4f} T, outside {B_out[2]:+.4f} T, jump {B_in[2] - B_out[2]:.4f} T   mu0 K_b = mu0 M = {MU0 * M:.4f} T")
    print(f"        normal B_r inside {B_in[0]:+.5f} T, outside {B_out[0]:+.5f} T   (continuous)")
# The flat end face at z = L/2 carries no surface current: B_z is normal there and continuous.
pts = np.array([[0.03, 0, L / 2 - dz], [0.03, 0, L / 2 + dz]])     # the sheet is 7 cm away here, so one loop spacing is enough
B_in, B_out = bs.wire_B(mid, dl, I, pts)
print(f"end face: normal B_z just inside {B_in[2]:.4f} T, just outside {B_out[2]:.4f} T   (no current on the end face, so no jump; the small difference is the smooth variation of B_z over 2 dz)")
```

### 3.7 Magnetic circuits

Iron with $\mu_r \sim 10^3$ confines magnetic flux almost as well as copper confines current, and the mathematics of a closed iron path with a winding on it maps exactly onto a DC circuit. Consider a ring of soft iron with mean path length $l$ and cross-section $A$, wound with $N$ turns carrying $I$. Ampere's law for $\mathbf{H}$ around the ring: $Hl = NI$. Inside the iron $B = \mu H$, and the flux through the cross-section is $\Phi = BA$. Combining,
$$\Phi = \frac{NI}{l/(\mu A)} \qquad\text{or}\qquad \mathcal{F} = \Phi\,\mathcal{R}, \quad \mathcal{F} = NI, \quad \mathcal{R} = \frac{l}{\mu A}.$$
Read it as Ohm's law: the magnetomotive force $\mathcal{F}$ (ampere-turns) drives a flux $\Phi$ (webers) through a reluctance $\mathcal{R}$ (A/Wb), and reluctances in series add, like resistances, because the same flux passes through each piece and the $H$ drops $H_il_i$ add up to $NI$ around the loop. The analogy holds as long as the flux stays inside the iron, which the refraction rule of section 3.6 guarantees for $\mu_r \gg 1$.

**Worked example 11 (a gapped ring).** Iron ring, $l_i = 0.30\,$m, $A = 4\,$cm$^2$, $\mu_r = 2000$, cut to leave an air gap $l_g = 2\,$mm, wound with $N = 500$ turns carrying $1\,$A. Reluctance of the iron: $\mathcal{R}_i = l_i/(\mu_0\mu_rA) = 0.30/(4\pi\times10^{-7}\times2000\times4\times10^{-4}) = 2.98\times10^5\,$A/Wb. Reluctance of the gap: $\mathcal{R}_g = l_g/(\mu_0A) = 0.002/(4\pi\times10^{-7}\times4\times10^{-4}) = 3.98\times10^6\,$A/Wb. The two-millimetre gap has thirteen times the reluctance of thirty centimetres of iron and holds $93\%$ of the total. Flux: $\Phi = 500/(4.28\times10^6) = 1.17\times10^{-4}\,$Wb, so $B = \Phi/A = 0.29\,$T in both the iron and the gap ($B_n$ is continuous across the gap faces). The $H$ drops: in the iron $H_i = B/\mu_0\mu_r = 116\,$A/m over $0.3\,$m gives $35$ ampere-turns; in the gap $H_g = B/\mu_0 = 2.3\times10^5\,$A/m over $2\,$mm gives $465$ ampere-turns; sum $500$. Without the gap the same winding would give $\Phi = 500/\mathcal{R}_i = 1.68\times10^{-3}\,$Wb and $B = 4.2\,$T, which iron cannot carry: the core would saturate at about $2.1\,$T. Gaps are often put in on purpose, to make the flux depend on geometry rather than on the non-linear iron.

```python
l_i, A, mu_r, l_g, N, I_coil = 0.30, 4e-4, 2000, 2e-3, 500, 1.0
R_i, R_g = l_i / (MU0 * mu_r * A), l_g / (MU0 * A)
Phi = N * I_coil / (R_i + R_g)
B = Phi / A
print(f"reluctances: iron {R_i:.3e} A/Wb, gap {R_g:.3e} A/Wb, gap share {R_g / (R_i + R_g):.1%}")
print(f"flux {Phi:.3e} Wb, B = {B:.3f} T")
print(f"H drops: iron {B / (MU0 * mu_r) * l_i:.1f} A-turns, gap {B / MU0 * l_g:.1f} A-turns, sum {B / (MU0 * mu_r) * l_i + B / MU0 * l_g:.1f} = N I = {N * I_coil}")
print(f"without the gap: B = {N * I_coil / R_i / A:.2f} T, above the saturation of iron (about 2.1 T)")
```
