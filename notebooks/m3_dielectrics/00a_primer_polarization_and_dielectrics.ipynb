## 2. The ideas

**Conductors and insulators.** A conductor has charges that are free to move through the whole body: in a metal, roughly one electron per atom belongs to no atom in particular. In an insulator (a dielectric, in the language of this chapter) every charge is attached to a molecule or an atom and can move only within it. Put a conductor in a field and charge flows until the field inside is zero (module 2). Put an insulator in a field and no charge flows anywhere, yet the material still responds: each molecule distorts a little, and the sum of all those small distortions changes the field.

**How a molecule responds.** A neutral atom has its positive nucleus at the centre of a negative electron cloud. In a field $\mathbf{E}$ the nucleus is pushed one way and the cloud the other, and they settle a tiny distance apart: the atom has become an *induced dipole*, a pair of equal and opposite charges separated by a small displacement, with the positive end on the side the field points towards. Some molecules (water is the standard example) are dipoles even with no field, because their charge is not distributed symmetrically; these *permanent dipoles* are oriented at random in the absence of a field and a field turns them, on average, part way into alignment. Both mechanisms produce the same macroscopic result: a net dipole moment pointing along $\mathbf{E}$ in every small volume of the material.

**Polarization.** The *polarization* $\mathbf{P}$ is the dipole moment per unit volume, averaged over a region large compared with a molecule and small compared with the body. It is a vector field like any other and it is the single quantity that describes the state of a polarized dielectric. Its units are coulomb per square metre, the same as surface charge density, and that is no accident.

**Bound charge.** Think of a uniformly polarized slab as stacks of tiny dipoles, head to tail. Inside, the positive end of each dipole sits next to the negative end of the next and they cancel. At the two faces of the slab there is nothing to cancel against: the top face is left with a layer of positive charge and the bottom with negative. That charge is called *bound* because it cannot leave the molecules it belongs to, but it is as real as any other charge: it produces a field by Coulomb's law and it attracts or repels other charges. If the polarization is not uniform the cancellation inside is incomplete and a volume density of bound charge is left behind as well. The two results of section 3.2 are $\sigma_b = \mathbf{P}\cdot\hat{\mathbf{n}}$ on a surface and $\rho_b = -\nabla\cdot\mathbf{P}$ in the volume, and the total bound charge of any neutral body is zero.

**The field of a polarized body.** Once the bound charges are known the field they produce is pure module 1: superposition of Coulomb fields from $\sigma_b$ and $\rho_b$. There is no new force law. A uniformly polarized sphere, for example, has $\sigma_b = P\cos\theta$ on its surface and that charge produces a uniform field $-\mathbf{P}/3\epsilon_0$ inside, pointing against the polarization. This is the *depolarizing field*: the material's own bound charge opposes the field that created it, which is why the field inside a dielectric is weaker than outside.

**The displacement field.** The total charge is free charge (the charge we put somewhere on purpose: on plates, on wires) plus bound charge (the charge the material produced in response). Gauss' law counts both, and that is awkward because the bound charge depends on the field we are trying to find. Writing $\rho_b = -\nabla\cdot\mathbf{P}$ and moving it to the left side defines a new field, $\mathbf{D} = \epsilon_0\mathbf{E} + \mathbf{P}$, whose divergence is the free charge alone: $\nabla\cdot\mathbf{D} = \rho_f$. The field $\mathbf{D}$ is bookkeeping, not a new force; its only job is to let us apply Gauss' law with the charges we actually control.

**Linear dielectrics.** For most materials in fields that are not enormous, the polarization is proportional to the field: $\mathbf{P} = \epsilon_0\chi_e\mathbf{E}$ with a dimensionless *susceptibility* $\chi_e$. Then $\mathbf{D} = \epsilon_0(1 + \chi_e)\mathbf{E} = \epsilon_r\epsilon_0\mathbf{E}$, where $\epsilon_r = 1 + \chi_e$ is the *relative permittivity* (also called the dielectric constant). A single number now characterises the material: air has $\epsilon_r \approx 1.0006$, paper about 3, glass about 5, water about 80. Filling a capacitor with a dielectric multiplies its capacitance by $\epsilon_r$, because the depolarizing field cancels part of the plate field and more charge is needed to reach the same voltage.

**Boundary conditions.** Where $\epsilon_r$ changes, the field changes abruptly. Two rules say how: the component of $\mathbf{E}$ tangential to the interface is continuous (from a small loop and the fact that $\nabla\times\mathbf{E} = 0$), and the component of $\mathbf{D}$ normal to the interface jumps by the free surface charge there, which for an interface between two insulators is zero (from a small pillbox and $\nabla\cdot\mathbf{D} = \rho_f$). Together they say field lines bend at a dielectric surface the way light bends at a glass surface, and they are the conditions a numerical solver must respect at every material interface.

**What the main notebooks do.** Notebook 01 puts a polarization field on a grid, takes its divergence to get $\rho_b$, recovers $\sigma_b$ from the step at a surface, and computes the field of a polarized sphere from its bound charge. Notebook 02 combines $\mathbf{D} = \epsilon\mathbf{E}$, $\mathbf{E} = -\nabla V$ and $\nabla\cdot\mathbf{D} = \rho_f$ into one equation, $\nabla\cdot(\epsilon\nabla V) = -\rho_f$, solves it on 2D and 3D grids, and measures the boundary conditions on the result. ### 3.1 The electric dipole

Two charges, $+q$ and $-q$, separated by a displacement $\mathbf{d}$ pointing from the negative charge to the positive one, form an electric dipole with *dipole moment*
$$\mathbf{p} = q\,\mathbf{d}.$$
The moment is the only thing the outside world sees once you are further away than a few times $d$: a small strong pair and a large weak pair with the same $q d$ are indistinguishable from far off. Units: coulomb metre.

**Worked example 1 (the potential far from a dipole).** Put $+q$ at $z = +d/2$ and $-q$ at $z = -d/2$. At a point at distance $r$ from the centre and polar angle $\theta$ from the $z$ axis, the distances to the two charges are $r_\pm = |\mathbf{r} \mp \tfrac12 d\,\hat{\mathbf{z}}|$. For $r \gg d$ the law of cosines gives $r_\pm \approx r \mp \tfrac12 d\cos\theta$, so
$$\frac{1}{r_\pm} \approx \frac1r\left(1 \pm \frac{d\cos\theta}{2r}\right).$$
The potential is the sum of two point-charge potentials (module 1, notebook 03),
$$V = \frac{q}{4\pi\epsilon_0}\left(\frac1{r_+} - \frac1{r_-}\right) \approx \frac{q}{4\pi\epsilon_0}\,\frac{d\cos\theta}{r^2} = \frac{p\cos\theta}{4\pi\epsilon_0 r^2} = \frac{\mathbf{p}\cdot\hat{\mathbf{r}}}{4\pi\epsilon_0 r^2}.$$
The $1/r$ terms cancelled, because the dipole is neutral; what survives falls as $1/r^2$, one power faster than a single charge. Taking $\mathbf{E} = -\nabla V$ in spherical coordinates (module 1, primer 00d):
$$E_r = -\frac{\partial V}{\partial r} = \frac{2p\cos\theta}{4\pi\epsilon_0 r^3}, \qquad E_\theta = -\frac1r\frac{\partial V}{\partial\theta} = \frac{p\sin\theta}{4\pi\epsilon_0 r^3}.$$
On the axis ($\theta = 0$) the field is $2p/4\pi\epsilon_0 r^3$ along $\mathbf{p}$; in the equatorial plane ($\theta = 90^\circ$) it is $p/4\pi\epsilon_0 r^3$ pointing *against* $\mathbf{p}$. The field falls as $1/r^3$. Keep the three powers in mind: charge $1/r^2$, dipole $1/r^3$, and the dipole potential $1/r^2$.

**Worked example 2 (a water molecule).** Water has a permanent dipole moment $p = 6.2\times10^{-30}\,\mathrm{C\,m}$. As an equivalent pair of elementary charges $e = 1.6\times10^{-19}\,\mathrm{C}$ that is a separation $d = p/e = 3.9\times10^{-11}\,\mathrm{m}$, about a tenth of the molecule's size. On the axis at $r = 1\,\mathrm{nm}$: $V = Kp/r^2 = 8.99\times10^9 \times 6.2\times10^{-30}/10^{-18} = 0.056\,\mathrm{V}$ and $E = 2Kp/r^3 = 1.1\times10^8\,\mathrm{V/m}$. Molecular fields are enormous by laboratory standards, which is why neighbouring molecules influence each other so strongly in a liquid.

**Worked example 3 (torque and energy in a uniform field).** In a uniform field $\mathbf{E}$ the two charges feel equal and opposite forces $\pm q\mathbf{E}$: no net force, but a torque $\boldsymbol{\tau} = \mathbf{p}\times\mathbf{E}$ of magnitude $pE\sin\theta$ that turns $\mathbf{p}$ towards $\mathbf{E}$. The potential energy is $U = -\mathbf{p}\cdot\mathbf{E}$, lowest when aligned. For water in $E = 10^5\,\mathrm{V/m}$ at $90^\circ$: $\tau = 6.2\times10^{-30}\times10^5 = 6.2\times10^{-25}\,\mathrm{N\,m}$, and the energy difference between aligned and anti-aligned is $2pE = 1.2\times10^{-24}\,\mathrm{J}$. Compare the thermal energy $k_BT = 4.1\times10^{-21}\,\mathrm{J}$ at room temperature: the alignment energy is three thousand times smaller. Thermal agitation keeps the molecules almost random, and the field wins only a slight statistical preference for alignment, proportional to $pE/k_BT$. That is why the polarization of a material is proportional to $E$ for ordinary fields: the response is a small perturbation of a random state. In a *non-uniform* field the two charges feel slightly different forces and a net force $(\mathbf{p}\cdot\nabla)\mathbf{E}$ appears, pulling the dipole towards stronger field. A charged comb picks up scraps of paper this way: it polarizes them and then attracts the polarization.

The code represents the water dipole as a pair of point charges, computes its potential with the module 1 superposition routine, and compares with the far-field formula at $1\,\mathrm{nm}$ (where $d/r = 0.04$ and the approximation is good) and at $0.1\,\mathrm{nm}$ (where it is not).


```python
p_water = 6.2e-30                                   # C m
e = 1.602e-19
d = p_water / e
print(f"equivalent separation d = p/e = {d:.2e} m")
src, q = di.dipole_sources([[0, 0, 0]], [[0, 0, p_water]], delta=d)

for r in (1e-9, 1e-10):
    theta = np.radians([0, 30, 60, 90, 135])
    obs = r * np.stack([np.sin(theta), 0 * theta, np.cos(theta)], axis=-1)
    V_pair = fields.point_charges_V(src, q, obs)
    V_far = K * p_water * np.cos(theta) / r**2
    print(f"\nr = {r:.0e} m  (d/r = {d / r:.2f})")
    print("  theta [deg]:         ", np.degrees(theta))
    print("  V of the pair [V]:   ", V_pair)
    print("  p cos(theta)/4 pi eps0 r^2:", V_far)

# Field on the axis and in the equatorial plane at 1 nm, against the spherical-component formulas.
r = 1e-9
E_axis = fields.point_charges_E(src, q, np.array([0, 0, r]))[2]
E_equator = fields.point_charges_E(src, q, np.array([r, 0, 0]))[2]
print(f"\nE_z on the axis at 1 nm:      {E_axis:.3e} V/m    formula 2Kp/r^3 = {2 * K * p_water / r**3:.3e}")
print(f"E_z in the equatorial plane:  {E_equator:.3e} V/m    formula -Kp/r^3 = {-K * p_water / r**3:.3e}")

# Worked example 3.
E_lab, kT = 1e5, 1.381e-23 * 300
print(f"\ntorque at 90 deg in {E_lab:.0e} V/m: {p_water * E_lab:.2e} N m;  2pE = {2 * p_water * E_lab:.2e} J;  k_B T = {kT:.2e} J;  ratio pE/kT = {p_water * E_lab / kT:.1e}")
```

```python
# The dipole field in the xz plane: unit arrows coloured by log10 |E|, and field lines.
xg = np.linspace(-2, 2, 41); X, Z = np.meshgrid(xg, xg, indexing="ij")
obs = np.stack([X, np.zeros_like(X), Z], axis=-1)
src_u, q_u = di.dipole_sources([[0, 0, 0]], [[0, 0, 1.0]], delta=0.05)          # unit moment, short pair
E = fields.point_charges_E(src_u, q_u, obs)
Emag = np.linalg.norm(E, axis=-1)
fig, axes = plt.subplots(1, 2, figsize=(12, 5.5))
st = 2
axes[0].quiver(X[::st, ::st], Z[::st, ::st], (E[..., 0] / Emag)[::st, ::st], (E[..., 2] / Emag)[::st, ::st], np.log10(Emag)[::st, ::st], cmap="plasma", pivot="mid")
axes[0].set_aspect("equal"); axes[0].set_title("dipole along z: unit arrows, colour = log10 |E|"); axes[0].set_xlabel("x"); axes[0].set_ylabel("z")
axes[1].streamplot(xg, xg, E[..., 0].T, E[..., 2].T, color=np.log10(Emag).T, cmap="plasma", density=1.4)
axes[1].set_aspect("equal"); axes[1].set_title("field lines leave +q and return to -q: along p on the axis, against p at the equator")
plt.show()
```

### 3.2 Polarization and bound charge

A dielectric with $N$ molecules per unit volume, each with average dipole moment $\langle\mathbf{p}\rangle$, has polarization
$$\mathbf{P} = N\langle\mathbf{p}\rangle, \qquad [\mathbf{P}] = \mathrm{C\,m}/\mathrm{m}^3 = \mathrm{C/m^2}.$$
$\mathbf{P}$ is a macroscopic field: it is defined by averaging over a volume that contains many molecules, so it varies smoothly even though the molecules do not. Everything in this chapter is about that averaged description.

**Why a polarized body carries charge: the counting argument.** Take a thin column of material along $\mathbf{P}$, of cross-section $A$ and length $L$, and picture it as a chain of dipoles laid head to tail. Every negative end inside the column sits against the positive end of the dipole below it, and the pair cancels. Only the top end of the column, with a positive charge, and the bottom end, with a negative charge, are left uncancelled. How much charge? The column contains total dipole moment $PAL$ (moment per volume times volume), and a chain that reduces to charges $\pm Q$ at its two ends a distance $L$ apart has moment $QL$. So $QL = PAL$, $Q = PA$, and the charge per unit area on the end faces is $\sigma_b = P$. If the end face is tilted so that its outward normal $\hat{\mathbf{n}}$ makes an angle with $\mathbf{P}$, only the component of $\mathbf{P}$ along the normal reaches the face, and
$$\sigma_b = \mathbf{P}\cdot\hat{\mathbf{n}}.$$

Now let $\mathbf{P}$ vary from place to place. A small box inside the material has bound charge $\mathbf{P}\cdot\hat{\mathbf{n}}$ on each face by the argument above, so the total charge *pushed out* of the box through its surface is $\oint\mathbf{P}\cdot d\mathbf{S}$, and the charge *left inside* is minus that. Divide by the volume and shrink (module 1, primer 00c): the bound charge per unit volume is
$$\rho_b = -\nabla\cdot\mathbf{P}.$$
Where $\mathbf{P}$ is uniform the divergence vanishes and there is no volume charge at all; a non-uniform $\mathbf{P}$ means more dipole charge leaves a region than enters it, and the difference stays behind. The formal route to the same pair of formulas writes the potential of the whole body as an integral of dipole potentials, $\int \mathbf{P}\cdot\hat{\mathbf{R}}/4\pi\epsilon_0R^2\,dv'$, notices that $\hat{\mathbf{R}}/R^2 = \nabla'(1/R)$, and integrates by parts: the surface term is the potential of $\sigma_b$ and the volume term is the potential of $\rho_b$. The counting argument is the physics; the integration by parts is the proof that the counting argument is exact.

**Bound charge is real charge, and it sums to zero.** Nothing distinguishes a bound charge from a free one in Coulomb's law. The only difference is mobility: bound charge cannot be drawn off through a wire because each piece of it is tethered to its molecule. For a body with no free charge, the divergence theorem applied to $\mathbf{P}$ gives $\int\rho_b\,dv + \oint\sigma_b\,dS = -\oint\mathbf{P}\cdot d\mathbf{S} + \oint\mathbf{P}\cdot d\mathbf{S} = 0$: polarization moves charge around inside a body, it never creates any.

**Worked example 4 (a uniformly polarized slab).** $\mathbf{P} = P\hat{\mathbf{z}}$ for $|z| < t$, zero outside, with $P = 10^{-6}\,\mathrm{C/m^2}$. Inside, $\nabla\cdot\mathbf{P} = \partial P/\partial z = 0$: no volume charge. Top face, $\hat{\mathbf{n}} = +\hat{\mathbf{z}}$: $\sigma_b = +P = +10^{-6}\,\mathrm{C/m^2}$. Bottom face, $\hat{\mathbf{n}} = -\hat{\mathbf{z}}$: $\sigma_b = -10^{-6}\,\mathrm{C/m^2}$. Side faces: $\hat{\mathbf{n}}\perp\mathbf{P}$, $\sigma_b = 0$. Total: zero. The slab is two parallel sheets of opposite charge, exactly the configuration of a parallel-plate capacitor, and the field between them (module 2, primer 00a) is $\sigma_b/\epsilon_0 = P/\epsilon_0$ pointing from $+$ to $-$, that is, from the top face down: against $\mathbf{P}$. Inside a uniformly polarized slab, $\mathbf{E} = -\mathbf{P}/\epsilon_0$ (for a slab much wider than it is thick).

**Worked example 5 (a uniformly polarized sphere).** $\mathbf{P} = P\hat{\mathbf{z}}$ inside radius $R$. No volume charge. On the surface $\hat{\mathbf{n}} = \hat{\mathbf{r}}$ and $\sigma_b = P\hat{\mathbf{z}}\cdot\hat{\mathbf{r}} = P\cos\theta$: positive on the northern hemisphere, negative on the southern, zero on the equator. Total: $\int_0^\pi P\cos\theta\cdot2\pi R^2\sin\theta\,d\theta = 0$. Dipole moment of the surface charge: $\int z\,\sigma_b\,dS = \int R\cos\theta\cdot P\cos\theta\cdot2\pi R^2\sin\theta\,d\theta = 2\pi PR^3\cdot\tfrac23 = \tfrac43\pi R^3P$, which is $P$ times the volume. The bound surface charge carries exactly the body's dipole moment, as it must.

**Worked example 6 (a radially polarized ball).** $\mathbf{P} = k\,r\,\hat{\mathbf{r}} = k(x, y, z)$ inside radius $R$. Volume charge: $\rho_b = -\nabla\cdot\mathbf{P} = -3k$ (the divergence of $(x, y, z)$ is 3), uniform and negative. Surface charge: $\sigma_b = \mathbf{P}\cdot\hat{\mathbf{r}}|_R = kR$, uniform and positive. Totals: volume $-3k\cdot\tfrac43\pi R^3 = -4\pi kR^3$, surface $kR\cdot4\pi R^2 = +4\pi kR^3$. Zero, as promised. This is the simplest example in which $\rho_b \ne 0$, and it shows the pattern: dipoles pointing outward and growing with $r$ push more charge out of each shell than into it.

The code puts the slab of worked example 4 on a grid. The discrete divergence cannot see a sharp surface; it sees $P_z$ drop from $P$ to zero across one or two cells and reports a spike of $\rho_b$ there. Integrating the spike through the surface recovers $\sigma_b = \pm P$. Notebook 01 uses exactly this. The second half builds the ball of worked example 6 and checks $\rho_b = -3k$ inside and a zero total.


```python
# Worked example 4 on a grid: a slab |z| < t, polarized along z.
P0, t = 1e-6, 0.5
x, y, z, G = grids.grid3d((-0.2, 0.2), (-0.2, 0.2), (-1.0, 1.0), (5, 5, 201))
h = grids.spacing(x, y, z)
Pol = np.zeros_like(G); Pol[..., 2] = np.where(np.abs(G[..., 2]) < t, P0, 0.0)
rho_b = di.bound_volume_charge(Pol, *h)
col = rho_b[2, 2, :]                                     # the centre column, a function of z
top = (z > 0.25) & (z < 0.75); bottom = (z > -0.75) & (z < -0.25)
print(f"rho_b in the middle of the slab: {col[100]:.1e} C/m^3   (hand: 0)")
print(f"integral of rho_b dz through the top face:    {np.sum(col[top]) * h[2]:+.3e} C/m^2   (hand: +P = {P0:+.1e})")
print(f"integral of rho_b dz through the bottom face: {np.sum(col[bottom]) * h[2]:+.3e} C/m^2   (hand: -P)")
print("sigma_b by the formula P . n:  top", di.bound_surface_charge(np.array([0, 0, P0]), [0, 0, 1]), " bottom", di.bound_surface_charge(np.array([0, 0, P0]), [0, 0, -1]))

fig, ax = plt.subplots(figsize=(7, 3.5))
ax.plot(z, Pol[2, 2, :, 2] / P0, label="P_z / P"); ax.plot(z, col * h[2] / P0, label="rho_b h / P  (spike at each face)")
ax.set_xlabel("z"); ax.legend(); ax.set_title("uniform polarization: bound charge only where P steps"); plt.show()

# Worked example 6 on a grid: P = k r r_hat inside radius R.
k_pol, R = 1e-6, 1.0
x, y, z, G = grids.grid3d((-1.5, 1.5), (-1.5, 1.5), (-1.5, 1.5), 61)
h = grids.spacing(x, y, z)
r = grids.magnitude(G)
Pol = np.where((r <= R)[..., None], k_pol * G, 0.0)
rho_b = di.bound_volume_charge(Pol, *h)
deep = r < 0.8 * R
print(f"\nradially polarized ball: rho_b deep inside = {rho_b[deep].mean():.3e} (spread {rho_b[deep].std():.1e})   hand: -3k = {-3 * k_pol:.3e}")
Q_vol_hand, Q_surf_hand = -3 * k_pol * 4 / 3 * np.pi * R**3, k_pol * R * 4 * np.pi * R**2
Q_total_grid = rho_b.sum() * h[0] * h[1] * h[2]
print(f"hand: volume charge {Q_vol_hand:+.3e} C, surface charge {Q_surf_hand:+.3e} C, total 0")
print(f"grid: total bound charge (volume and the smeared surface spike together) = {Q_total_grid:+.1e} C")
```

### 3.3 The field of a polarized body

Having replaced the dielectric by its bound charges, the field it produces is the ordinary superposition field of those charges. No new law is involved. For the uniformly polarized sphere there is a closed form worth knowing because it is the model for every dielectric problem that follows.

**The two displaced balls.** A uniformly polarized sphere is the same thing as two uniformly charged balls, one of density $+\rho$ centred at $+\tfrac12\mathbf{d}$ and one of density $-\rho$ centred at $-\tfrac12\mathbf{d}$, with $\rho\,\mathbf{d} = \mathbf{P}$: where the two balls overlap the charge cancels, and the thin crescents left uncovered at the surface are exactly the $P\cos\theta$ surface charge. Inside a uniformly charged ball the field is $\rho\,\mathbf{r}_c/3\epsilon_0$ with $\mathbf{r}_c$ measured from the ball's centre (module 2, primer 00a, worked example 4). Add the two:
$$\mathbf{E}_{\text{in}} = \frac{\rho(\mathbf{r} - \tfrac12\mathbf{d})}{3\epsilon_0} - \frac{\rho(\mathbf{r} + \tfrac12\mathbf{d})}{3\epsilon_0} = -\frac{\rho\,\mathbf{d}}{3\epsilon_0} = -\frac{\mathbf{P}}{3\epsilon_0}.$$
Uniform, and against $\mathbf{P}$. The $\mathbf{r}$ cancelled, which is why the interior field is the same at every point. Outside, each ball looks like a point charge at its centre (Gauss), so the pair looks like a point dipole of moment $p = \rho\,d\cdot\tfrac43\pi R^3 = \tfrac43\pi R^3P$ at the origin, with the dipole field of section 3.1.

The factor $\tfrac13$ is specific to the sphere. For the slab of worked example 4 it was $1$ ($\mathbf{E} = -\mathbf{P}/\epsilon_0$), and for a long cylinder polarized across its axis it is $\tfrac12$. These *depolarizing factors* measure how much of a body's own bound charge the interior sees, and they are what makes the field inside a dielectric depend on its shape.

**Worked example 7.** A sphere of radius $R = 1\,\mathrm{m}$ with $P = 10^{-6}\,\mathrm{C/m^2}$. Inside: $E = -P/3\epsilon_0 = -10^{-6}/(3\times8.854\times10^{-12}) = -3.76\times10^4\,\mathrm{V/m}$ along $z$. Dipole moment $p = \tfrac43\pi R^3P = 4.19\times10^{-6}\,\mathrm{C\,m}$. On the axis at $z = 2R$: $E_z = 2Kp/z^3 = 2\times8.99\times10^9\times4.19\times10^{-6}/8 = 9.41\times10^3\,\mathrm{V/m}$, now *along* $\mathbf{P}$. In the equatorial plane at $r = 2R$: $E_z = -Kp/r^3 = -4.71\times10^3\,\mathrm{V/m}$. Note the sign change at the surface on the axis: just inside the north pole the field is $-3.76\times10^4$, just outside it is $+2Kp/R^3 = +7.53\times10^4$. The jump is $\sigma_b/\epsilon_0 = P/\epsilon_0 = 1.13\times10^5\,\mathrm{V/m}$, the discontinuity a sheet of charge always produces (module 2, primer 00a).

The code covers the sphere with 20 000 equal-area patches carrying $\sigma_b = P\cos\theta$ and evaluates the superposition field at interior and exterior points.


**Animation.** The two-displaced-balls argument in motion. A $+\rho$ ball and a $-\rho$ ball start on top of each other, neutral everywhere. Sliding them apart by $d$ leaves the grey overlap neutral and uncovers two crescents of thickness $d\cos\theta$, which are the surface charge $P\cos\theta$. In the last part $d$ shrinks while $\rho$ grows with $P = \rho d$ fixed: the crescents become a surface layer and the interior field does not change.

```python
def sphere_patches(R, N):
    '''Fibonacci-spiral patches: centres (N, 3), outward unit normals (N, 3), and the equal area per patch.'''
    k = np.arange(N) + 0.5
    theta = np.arccos(1 - 2 * k / N); phi = np.pi * (1 + 5**0.5) * k
    n = np.stack([np.sin(theta) * np.cos(phi), np.sin(theta) * np.sin(phi), np.cos(theta)], axis=-1)
    return R * n, n, 4 * np.pi * R**2 / N

R, P0 = 1.0, 1e-6
centres, normals, dA = sphere_patches(R, 20000)
sigma_b = P0 * normals[:, 2]                              # P . n = P cos(theta) on every patch
q_patch = sigma_b * dA
print(f"total bound charge of the sphere: {q_patch.sum():.1e} C   (hand: 0)")
print(f"dipole moment of the patches: {np.sum(q_patch * centres[:, 2]):.4e} C m   (hand: 4/3 pi R^3 P = {4 / 3 * np.pi * R**3 * P0:.4e})")

E_in_hand = -P0 / (3 * EPS0)
obs_in = np.array([[0, 0, 0], [0.3, 0.1, -0.2], [-0.5, 0.4, 0.3], [0, 0, 0.8]])
E_in = fields.point_charges_E(centres, q_patch, obs_in)
print(f"\ninterior field, hand: E = ({0:.0f}, {0:.0f}, {E_in_hand:.4e}) V/m at every point")
for p, e in zip(obs_in, E_in):
    print(f"  at {p}: E = {e}")

p_sphere = 4 / 3 * np.pi * R**3 * P0
for name, pt, hand in (("axis, z = 2R", [0, 0, 2.0], 2 * K * p_sphere / 8), ("equator, r = 2R", [2.0, 0, 0], -K * p_sphere / 8)):
    e = fields.point_charges_E(centres, q_patch, np.array(pt))
    print(f"outside on the {name}: E_z = {e[2]:.4e} V/m   dipole formula {hand:.4e}")
```

**Macroscopic and microscopic fields.** The bound-charge route gives the *macroscopic* field: the field averaged over a region containing many molecules, which is the field that appears in every formula of this chapter. If instead you fill the sphere with a lattice of point dipoles and add up their fields at one point, you get the *microscopic* field, which swings wildly from place to place: at a lattice site of a cubic array it is nearly zero (the contributions of the surrounding dipoles cancel by symmetry, a result associated with Lorentz), between sites it can be large in either direction. Averaged over a cell the microscopic field reproduces $-\mathbf{P}/3\epsilon_0$, but any single sample of it does not. Notebook 01 explores this directly. The lesson for computation: to get the field of a dielectric, compute the field of its bound charge densities, never the sum over individual dipoles.


### 3.4 The displacement field and the generalized Gauss' law

Inside matter, Gauss' law still holds with the *total* charge density: $\nabla\cdot\mathbf{E} = (\rho_f + \rho_b)/\epsilon_0$, where $\rho_f$ is the *free* charge, placed deliberately (on conductors, in an electron beam, as ions in a solution), and $\rho_b$ is the bound charge the material produces in response. The difficulty is that $\rho_b = -\nabla\cdot\mathbf{P}$ depends on $\mathbf{E}$ through the material's response, so we cannot apply Gauss' law with a known right-hand side. Substitute and rearrange:
$$\epsilon_0\nabla\cdot\mathbf{E} = \rho_f - \nabla\cdot\mathbf{P} \quad\Longrightarrow\quad \nabla\cdot(\epsilon_0\mathbf{E} + \mathbf{P}) = \rho_f.$$
Define the *electric displacement*
$$\mathbf{D} = \epsilon_0\mathbf{E} + \mathbf{P}, \qquad \nabla\cdot\mathbf{D} = \rho_f, \qquad \oint_S\mathbf{D}\cdot d\mathbf{S} = Q_{f,\text{enc}}.$$
This is Gauss' law with only the free charge on the right, and the integral form follows from the divergence theorem as in module 2. Every symmetry argument from primer M2.0a applies to $\mathbf{D}$ unchanged: for a charged ball, line, or sheet of free charge surrounded by *any* arrangement of dielectric with the same symmetry, $\mathbf{D}$ is what the vacuum formula gave for $\epsilon_0\mathbf{E}$, and the dielectric cannot change it.

**What $\mathbf{D}$ is and is not.** $\mathbf{D}$ has the units of $\mathbf{P}$, coulomb per square metre, and it is a device for bookkeeping free charge. It exerts no force; the force on a charge is always $q\mathbf{E}$. It does not satisfy a Coulomb-like law in general: $\nabla\times\mathbf{D} = \nabla\times\mathbf{P}$, which is not zero when $\mathbf{P}$ curls (for instance at the edge of a polarized slab), so $\mathbf{D}$ is *not* determined by its divergence alone and there is no "$\mathbf{D}$ potential". In the symmetric cases, and in linear materials where $\mathbf{P}\parallel\mathbf{E}$, the curl vanishes and $\mathbf{D}$ behaves well; that covers everything in this module.

**Worked example 8 (a parallel-plate capacitor with a dielectric slab, by pillbox).** Plates carry free charge $\pm\sigma_f = \pm10^{-6}\,\mathrm{C/m^2}$, separation $1\,\mathrm{mm}$, and a slab with $\epsilon_r = 4$ of thickness $0.5\,\mathrm{mm}$ lies against the negative plate, leaving a $0.5\,\mathrm{mm}$ vacuum gap at the positive plate. Pillbox with one face inside the positive plate's metal (where every field is zero) and the other face anywhere in the gap or in the slab: enclosed free charge $\sigma_fA$, so $D = \sigma_f = 10^{-6}\,\mathrm{C/m^2}$ everywhere between the plates, in the gap and in the slab alike. In the gap, $\mathbf{P} = 0$ and $E = D/\epsilon_0 = 1.13\times10^5\,\mathrm{V/m}$. In the slab (anticipating section 3.5, $\mathbf{D} = \epsilon_r\epsilon_0\mathbf{E}$), $E = D/\epsilon_r\epsilon_0 = 2.82\times10^4\,\mathrm{V/m}$, four times smaller. The polarization in the slab is $P = D - \epsilon_0E = \sigma_f(1 - 1/\epsilon_r) = 7.5\times10^{-7}\,\mathrm{C/m^2}$, and the bound surface charge on the slab's face towards the positive plate is $-P$: that negative layer cancels three quarters of the plate's field inside the slab. The voltage between the plates is $E_{\text{gap}}\cdot0.5\,\mathrm{mm} + E_{\text{slab}}\cdot0.5\,\mathrm{mm} = 56.5 + 14.1 = 70.6\,\mathrm{V}$, against $112.9\,\mathrm{V}$ for the empty capacitor: the same charge produces less voltage, so the capacitance has risen by $112.9/70.6 = 1.6$.

The code solves this problem on a grid with `di.solve_dielectric`, which notebook 02 builds on, and reads $E$, $D$ and the bound charge off the solution. One detail: the solver puts $\epsilon_r$ on nodes and uses the harmonic mean on the faces between them, so the interface it represents sits halfway between the last slab node and the first gap node; the code positions the slab accordingly.


```python
def slab_capacitor(n, eps_r, fill):
    '''Plates at x = 0 (V = 0) and x = 1 (V = 1), dielectric on nodes with x < fill. Side edges hold the exact 1D solution.
    The discrete interface sits halfway between the last slab node and the first vacuum node (x_int).'''
    x = np.linspace(0, 1, n)
    slab = x < fill
    k = np.argmax(~slab)
    x_int = 0.5 * (x[k - 1] + x[k])
    D_over_eps0 = 1.0 / (x_int / eps_r + (1 - x_int))
    exact = np.where(x < x_int, D_over_eps0 * x / eps_r, D_over_eps0 * (x_int / eps_r + (x - x_int)))
    V0 = np.zeros((n, n)); V0[:, 0] = V0[:, -1] = exact; V0[-1, :] = 1.0
    fixed = np.zeros((n, n), dtype=bool)
    er = np.where(slab[:, None], eps_r, 1.0) * np.ones((1, n))
    return x, exact, V0, fixed, er, x_int

n, eps_r = 81, 4.0
x, exact, V0, fixed, er, x_int = slab_capacitor(n, eps_r, fill=0.5)
h = x[1] - x[0]
V = di.solve_dielectric(V0, fixed, er, h=h)
E_line = np.abs(-np.gradient(V[:, n // 2], h))            # |E_x| along the centre line; E points from the 1 V plate to the 0 V plate, so E_x < 0
D_line = EPS0 * er[:, n // 2] * E_line
k = int(round(x_int / h))
E_slab, E_gap = E_line[3:k - 3].mean(), E_line[k + 4:-3].mean()
print(f"slab on x < {x_int:.4f}: E in slab {E_slab:.4f} V/m, E in gap {E_gap:.4f} V/m, ratio {E_gap / E_slab:.4f}  (hand: eps_r = {eps_r})")
print(f"D in slab {D_line[3:k - 3].mean():.4e}, D in gap {D_line[k + 4:-3].mean():.4e}  (same value: D is continuous, no free charge at the interface)")
sigma_f = D_line[k + 4:-3].mean()
P_slab = sigma_f * (1 - 1 / eps_r)
print(f"free charge on the plates sigma_f = D = {sigma_f:.4e} C/m^2;  P in the slab = D - eps0 E = {sigma_f - EPS0 * E_slab:.4e}  (hand: sigma_f (1 - 1/eps_r) = {P_slab:.4e})")
print(f"max |V - exact 1D solution| on the grid: {np.abs(V - exact[:, None]).max():.1e}")

# Worked example 8 with its numbers.
sf, d_gap, d_slab = 1e-6, 0.5e-3, 0.5e-3
Eg, Es = sf / EPS0, sf / (eps_r * EPS0)
print(f"\nworked example 8: D = {sf:.1e} C/m^2, E_gap = {Eg:.3e} V/m, E_slab = {Es:.3e} V/m, P_slab = {sf * (1 - 1 / eps_r):.2e} C/m^2")
print(f"  voltage with slab {Eg * d_gap + Es * d_slab:.1f} V, empty {Eg * (d_gap + d_slab):.1f} V, capacitance ratio {Eg * (d_gap + d_slab) / (Eg * d_gap + Es * d_slab):.3f}")

fig, ax = plt.subplots(figsize=(7, 3.5))
ax.plot(x, E_line / E_line.max(), label="E / E_gap"); ax.plot(x, D_line / D_line.max(), label="D / D_gap")
ax.axvline(x_int, color="grey", ls=":"); ax.set_xlabel("x (from the 0 V plate to the 1 V plate)"); ax.legend(); ax.set_title("slab for x < 0.5: E drops by eps_r, D does not"); plt.show()
```

### 3.5 Linear dielectrics

Worked example 3 explained why the response of ordinary matter to ordinary fields is proportional to the field. Write that proportionality as
$$\mathbf{P} = \epsilon_0\chi_e\mathbf{E},$$
where $\chi_e$, the *electric susceptibility*, is a dimensionless property of the material ($\epsilon_0$ is pulled out so that $\chi_e$ is a pure number). Then
$$\mathbf{D} = \epsilon_0\mathbf{E} + \epsilon_0\chi_e\mathbf{E} = \epsilon_0(1 + \chi_e)\mathbf{E} = \epsilon_r\epsilon_0\mathbf{E} = \epsilon\mathbf{E},$$
with $\epsilon_r = 1 + \chi_e$ the *relative permittivity* and $\epsilon = \epsilon_r\epsilon_0$ the *permittivity*. Vacuum has $\epsilon_r = 1$ exactly; air $1.0006$; paper and most plastics 2 to 4; glass 4 to 7; water 80 (its permanent dipoles are easy to turn); some ceramics several thousand. The number is what you look up for a material, and in the main notebooks it is the array `eps_r` on the grid.

Materials that obey this relation are *linear* (double the field, double the polarization) and, if $\epsilon_r$ is the same in every direction, *isotropic*. Crystals can be anisotropic, with $\epsilon_r$ a matrix, and strong fields eventually break linearity; neither case appears in this course.



**Worked example 9 (a filled capacitor).** Plates of area $A = 1\,\mathrm{cm^2}$ and separation $d = 1\,\mathrm{mm}$: $C_0 = \epsilon_0A/d = 8.854\times10^{-12}\times10^{-4}/10^{-3} = 0.885\,\mathrm{pF}$. Fill the gap with glass, $\epsilon_r = 4$. By the pillbox argument of worked example 8, $D = \sigma_f$ is unchanged and $E = \sigma_f/\epsilon_r\epsilon_0$ is four times smaller, so $V = Ed$ is four times smaller for the same charge: $C = Q/V = \epsilon_rC_0 = 3.54\,\mathrm{pF}$. In general, for any capacitor completely filled with a linear dielectric, $C = \epsilon_rC_0$. Two ways to think about it: at fixed charge (plates disconnected) the dielectric lowers the voltage by $\epsilon_r$; at fixed voltage (battery connected) the battery pushes $\epsilon_r$ times more charge onto the plates to restore the voltage the bound charge cancelled.

**Worked example 10 (a partially filled capacitor).** A slab of thickness $xd$ fills a fraction $x$ of the gap. $D = \sigma_f$ throughout, so $V = \sigma_f\left(\dfrac{xd}{\epsilon_r\epsilon_0} + \dfrac{(1-x)d}{\epsilon_0}\right)$ and
$$\frac{C}{C_0} = \frac{1}{x/\epsilon_r + (1 - x)}.$$
The gap and the slab act as two capacitors in series (module 2, primer 00c): same charge, voltages add. For $x = 0.5$ and $\epsilon_r = 4$: $C/C_0 = 1/(0.125 + 0.5) = 1.6$, as worked example 8 found. The limits check: $x = 0$ gives 1, $x = 1$ gives $\epsilon_r$. The slab's position in the gap does not matter, only its thickness. (A slab filling a fraction $y$ of the plate *area* instead, side by side with vacuum, gives two capacitors in parallel and $C/C_0 = (1 - y) + y\epsilon_r$.)

**Worked example 11 (energy).** The field energy density in a linear dielectric is $u = \tfrac12\mathbf{D}\cdot\mathbf{E} = \tfrac12\epsilon_r\epsilon_0E^2$: the vacuum expression with $\epsilon_0$ replaced by $\epsilon$. It is the work done to assemble the free charge against the field, with the energy of polarizing the molecules included. The filled capacitor of worked example 9 at $V = 100\,\mathrm{V}$ stores $U = \tfrac12CV^2 = \tfrac12\times3.54\times10^{-12}\times10^4 = 1.77\times10^{-8}\,\mathrm{J}$, four times the empty capacitor at the same voltage. At the same *charge* it stores $Q^2/2C$, four times *less* than empty: inserting the slab into a charged, disconnected capacitor releases energy, which is the work the field does pulling the slab in. A dielectric is always attracted into the region of strongest field, the same force that worked example 3 found for a single dipole.

The code computes the capacitance of the half-filled capacitor from the solver's energy, $C = 2U/V^2$, which is the route notebook 02 uses for geometries with no closed form. `di.energy` integrates $\tfrac12\epsilon_r\epsilon_0E^2$ on the faces between nodes with the solver's own face permittivity, so it reproduces the series formula to rounding error; a node-based integral would be off by a few parts in a thousand because of the interface node.


**Animation.** Two capacitors with the same free charge on their plates. On the left the gap is empty and the field is $E_0$. On the right a dielectric fills the gap: its molecules turn to align with the field, bound charge appears on the two faces of the slab, and the field inside drops to $E_0/\epsilon_r$. The voltage drops by the same factor, which is why the capacitance rises by $\epsilon_r$.

```python
V0_plates = 1.0
U = di.energy(V, er, h)                                   # per unit depth; plates of width 1, separation 1
C_num = 2 * U / V0_plates**2
C_0 = EPS0 * 1.0 / 1.0                                    # eps0 * width / separation, per unit depth
C_series = 1 / (x_int / (eps_r * EPS0) + (1 - x_int) / EPS0)
print(f"energy per unit depth U = {U:.6e} J/m, so C = 2U/V^2 = {C_num:.6e} F/m")
print(f"series formula with the slab on x < {x_int:.4f}: {C_series:.6e} F/m   relative difference {abs(C_num / C_series - 1):.1e}")
print(f"C / C_0 = {C_num / C_0:.4f}   (hand, worked example 10 with x = {x_int:.4f}: {1 / (x_int / eps_r + 1 - x_int):.4f})")

# Node-based energy for comparison: one permittivity per node.
E_node = -fields.gradient(V, h, h)
u_node = 0.5 * EPS0 * er * np.sum(E_node**2, axis=-1)
w = np.ones(n); w[0] = w[-1] = 0.5
U_node = np.sum(u_node * w[:, None] * w[None, :]) * h * h
print(f"node-based energy integral gives C / C_series = {2 * U_node / C_series:.4f}: first-order error at the interface node")

# Worked examples 9 and 11.
A, d_plate = 1e-4, 1e-3
C0 = EPS0 * A / d_plate
print(f"\nworked example 9: C_0 = {C0 * 1e12:.3f} pF, filled with eps_r = 4: {4 * C0 * 1e12:.3f} pF")
print(f"worked example 11: energy at 100 V, filled {0.5 * 4 * C0 * 100**2:.3e} J, empty {0.5 * C0 * 100**2:.3e} J")
```

### 3.6 Boundary conditions, and the dielectric sphere

At a surface where $\epsilon_r$ jumps, the fields jump too, and two rules fix how. Both come from applying the integral laws to a region that straddles the surface and shrinking it.

**Tangential $\mathbf{E}$ is continuous.** The electrostatic field has zero curl, so its line integral around any closed loop is zero (module 1, primer 00d). Take a thin rectangular loop with its two long sides of length $\ell$ parallel to the interface, one on each side, and its short sides negligible. The line integral is $E_{1t}\ell - E_{2t}\ell = 0$, so
$$E_{1t} = E_{2t}.$$
A tangential field that jumped would make a loop integral non-zero and a potential impossible. Equivalently, $V$ itself is continuous across the surface.

**Normal $\mathbf{D}$ jumps by the free surface charge.** Apply the generalized Gauss' law to a thin pillbox of face area $A$ straddling the surface: the flux is $D_{2n}A - D_{1n}A$ (normal pointing from 1 into 2) and the enclosed free charge is $\sigma_fA$, so
$$D_{2n} - D_{1n} = \sigma_f.$$
Between two insulators there is no free charge and $D_n$ is continuous: $\epsilon_1E_{1n} = \epsilon_2E_{2n}$. The normal component of $\mathbf{E}$ therefore *jumps* by the factor $\epsilon_1/\epsilon_2$, and the jump is produced by the bound surface charge: $E_{2n} - E_{1n} = \sigma_b/\epsilon_0$, the same discontinuity any sheet of charge makes.

**Worked example 12 (refraction of field lines).** A field in air ($\epsilon_1 = 1$) meets a glass surface ($\epsilon_2 = 4$) at $30^\circ$ to the normal. Tangential: $E_2\sin\theta_2 = E_1\sin\theta_1$. Normal: $4E_2\cos\theta_2 = E_1\cos\theta_1$. Divide: $\tan\theta_2 = 4\tan\theta_1 = 4\times0.577 = 2.31$, so $\theta_2 = 66.6^\circ$. The lines bend *away* from the normal on entering the denser medium, the opposite of light. The magnitude: $E_2 = E_1\sin30^\circ/\sin66.6^\circ = 0.545\,E_1$. In general $\tan\theta_1/\tan\theta_2 = \epsilon_1/\epsilon_2$.



**Worked example 13 (a dielectric sphere in a uniform field, by self-consistency).** A sphere of linear dielectric $\epsilon_r$ in an applied uniform field $\mathbf{E}_0$. Guess that the polarization inside comes out uniform. Then section 3.3 says the bound charge produces the uniform field $-\mathbf{P}/3\epsilon_0$ inside, so the total interior field is
$$\mathbf{E}_{\text{in}} = \mathbf{E}_0 - \frac{\mathbf{P}}{3\epsilon_0}.$$
But the material's own law says $\mathbf{P} = \epsilon_0(\epsilon_r - 1)\mathbf{E}_{\text{in}}$. Substitute:
$$\mathbf{E}_{\text{in}} = \mathbf{E}_0 - \frac{\epsilon_r - 1}{3}\mathbf{E}_{\text{in}} \quad\Longrightarrow\quad \mathbf{E}_{\text{in}} = \frac{3}{\epsilon_r + 2}\,\mathbf{E}_0.$$
Uniform, so the guess was consistent, and by the uniqueness theorem of module 2 it is *the* solution. For $\epsilon_r = 3$: $E_{\text{in}} = 0.6\,E_0$ and $P = \epsilon_0\cdot2\cdot0.6E_0 = 1.2\,\epsilon_0E_0$. Outside, the sphere adds the field of a dipole $p = \tfrac43\pi R^3P = 4\pi\epsilon_0R^3\,\dfrac{\epsilon_r - 1}{\epsilon_r + 2}E_0$ to the applied field, so with $k = (\epsilon_r-1)/(\epsilon_r+2)$
$$V_{\text{out}} = -E_0z + \frac{kR^3E_0z}{r^3}, \qquad V_{\text{in}} = -\frac{3}{\epsilon_r + 2}E_0z.$$
Check the boundary conditions at the pole ($z = R$ on the axis, everything normal): inside $D_n = \epsilon_r\epsilon_0\cdot3E_0/(\epsilon_r + 2)$; outside $D_n = \epsilon_0E_0(1 + 2k) = \epsilon_0E_0\cdot3\epsilon_r/(\epsilon_r + 2)$. Equal. Notebook 02 uses $V_{\text{out}}$ as the boundary value on a 3D grid and measures both conditions on the numerical solution.

The same argument for a long dielectric *cylinder* in a transverse field uses the depolarizing factor $\tfrac12$ and gives $E_{\text{in}} = \dfrac{2}{\epsilon_r + 1}E_0$ with $V_{\text{out}} = -E_0x + \dfrac{\epsilon_r - 1}{\epsilon_r + 1}\dfrac{R^2E_0x}{r^2}$. The code solves that 2D problem with `di.solve_dielectric`, fixing the outer edge of the grid to $V_{\text{out}}$, and compares the interior field with the closed form.


```python
eps_r, R, E0, n = 3.0, 1.0, 1.0, 121
L = 3.0 * R
x, y, P2 = grids.grid2d((-L, L), (-L, L), n)
h = x[1] - x[0]
X, Y = P2[..., 0], P2[..., 1]
r2 = np.hypot(X, Y)
k_cyl = (eps_r - 1) / (eps_r + 1)
with np.errstate(divide="ignore", invalid="ignore"):
    V_out = -E0 * X + k_cyl * R**2 * E0 * X / r2**2
V0 = np.where(r2 > R, V_out, 0.0)                          # only the outer edge of the grid is used as a boundary value
er = np.where(r2 <= R, eps_r, 1.0)
Vc = di.solve_dielectric(V0, np.zeros((n, n), dtype=bool), er, h=h)
Ec = -fields.gradient(Vc, h, h)
deep = r2 < 0.6 * R
E_in_num, E_in_hand = Ec[..., 0][deep].mean(), 2 / (eps_r + 1) * E0
print(f"dielectric cylinder, eps_r = {eps_r}: E_x deep inside = {E_in_num:.4f} E0 (spread {Ec[..., 0][deep].std():.1e}), hand 2/(eps_r + 1) = {E_in_hand:.4f};  E_y there = {Ec[..., 1][deep].mean():.1e}")
print(f"sphere for comparison (worked example 13): E_in = 3/(eps_r + 2) E0 = {3 / (eps_r + 2):.4f} E0")

# Refraction, worked example 12.
th1 = np.radians(30.0); th2 = np.arctan(4 * np.tan(th1))
print(f"\nworked example 12: theta_2 = {np.degrees(th2):.1f} deg,  E_2 / E_1 = {np.sin(th1) / np.sin(th2):.3f}")

fig, axes = plt.subplots(1, 2, figsize=(12, 5.5))
axes[0].contour(X, Y, Vc, levels=25, cmap="viridis"); axes[0].add_patch(plt.Circle((0, 0), R, fill=False, color="C3", lw=2))
axes[0].set_aspect("equal"); axes[0].set_title("equipotentials: closer outside, wider inside (E is weaker inside)")
speed = np.hypot(Ec[..., 0], Ec[..., 1])
axes[1].streamplot(x, y, Ec[..., 0].T, Ec[..., 1].T, color=speed.T, cmap="plasma", density=1.3)
axes[1].add_patch(plt.Circle((0, 0), R, fill=False, color="w", lw=2)); axes[1].set_aspect("equal"); axes[1].set_title("field lines bend at the surface, colour = |E|")
plt.show()
```

## 4. Verify

Two closed-form checks with asserts. First, the interior field of the uniformly polarized sphere built from its bound surface charge must be $-P/3\epsilon_0$ along $\mathbf{P}$, with no transverse component, at several interior points. Second, the total bound charge of the radially polarized ball on the grid must vanish: the smeared surface spike must cancel the uniform volume charge to the accuracy of the discrete divergence.


```python
E_in_num = fields.point_charges_E(centres, q_patch, np.array([[0, 0, 0], [0.2, -0.3, 0.1], [-0.4, 0.2, -0.5], [0.0, 0.0, 0.7]]))
err = np.abs(E_in_num[:, 2] / (-P0 / (3 * EPS0)) - 1).max()
trans = np.abs(E_in_num[:, :2]).max() / (P0 / (3 * EPS0))
print(f"polarized sphere: max relative error of E_z inside {err:.1e}, max transverse / |E| {trans:.1e}")
assert err < 1e-2 and trans < 1e-2

Q_scale = k_pol * R * 4 * np.pi * R**2                     # the surface charge magnitude, for scale
print(f"radially polarized ball: |total bound charge| / surface charge = {abs(Q_total_grid) / Q_scale:.1e}")
assert abs(Q_total_grid) < 1e-6 * Q_scale
print("verify passed")
```

| bound surface charge | $\sigma_b = \mathbf{P}\cdot\hat{\mathbf{n}}$ | $\hat{\mathbf{n}}$ outward |
| bound volume charge | $\rho_b = -\nabla\cdot\mathbf{P}$ | always |
| total bound charge | $\int\rho_b\,dv + \oint\sigma_b\,dS = 0$ | any body |

| uniformly polarized sphere | $\sigma_b = P\cos\theta$, $\mathbf{E}_{\text{in}} = -\mathbf{P}/3\epsilon_0$, dipole $p = \tfrac43\pi R^3P$ outside | always |

| generalized Gauss' law | $\nabla\cdot\mathbf{D} = \rho_f$, $\oint\mathbf{D}\cdot d\mathbf{S} = Q_{f,\text{enc}}$ | always |

## 7. Warm-up answers

1. $\mathbf{p} = q\mathbf{d}$ with $\mathbf{d}$ pointing from $-q$ to $+q$; the moment points from negative to positive.
2. The potential falls as $1/r^2$ and the field as $1/r^3$, one power faster than a single charge ($1/r$ and $1/r^2$), because the two charges' leading terms cancel.
3. Each molecule becomes, or turns into, a small dipole aligned with the field: electron clouds shift against nuclei, and permanent dipoles acquire a slight average alignment. Inside, the dipoles' charges cancel neighbour to neighbour; on the two faces where the field enters and leaves, a layer of uncancelled bound charge appears, negative where the field enters and positive where it leaves.
4. $\mathbf{P}$ is the dipole moment per unit volume, in $\mathrm{C/m^2}$. $\sigma_b = \mathbf{P}\cdot\hat{\mathbf{n}}$ with $\hat{\mathbf{n}}$ the outward normal; $\rho_b = -\nabla\cdot\mathbf{P}$.
5. Yes: it produces fields and forces by Coulomb's law like any other charge. No: it cannot be drawn off through a wire, because each piece of it is attached to a molecule and only displaced within it.
6. $\mathbf{D} = \epsilon_0\mathbf{E} + \mathbf{P}$. Its sources are the free charges only, $\nabla\cdot\mathbf{D} = \rho_f$. It is convenient because the free charge is what we control, while the bound charge depends on the unknown field; with $\mathbf{D}$ the symmetry arguments of Gauss' law go through without knowing the material's response in advance.
7. The tangential component of $\mathbf{E}$ (from the zero curl of $\mathbf{E}$, via a small loop) and the normal component of $\mathbf{D}$ (from $\nabla\cdot\mathbf{D} = \rho_f$ with $\sigma_f = 0$, via a small pillbox).
